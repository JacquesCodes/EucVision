```python
# @title Height Calibration

# ------------------------------------------------------------------------------
# 1. SETUP AND IMPORTS
# ------------------------------------------------------------------------------
# Mount google drive before running if using Google Colab.

import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import matplotlib.dates as mdates
import numpy as np
import scipy.stats as stats
import math
from matplotlib.colors import LinearSegmentedColormap
from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler
from matplotlib.patches import Ellipse
import matplotlib.transforms as transforms
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import cross_val_predict
from sklearn.model_selection import KFold
from sklearn.metrics import r2_score, mean_squared_error, mean_absolute_error
import matplotlib.patches as mpatches
from statsmodels.nonparametric.smoothers_lowess import lowess
import matplotlib.dates as mdates
import math
import matplotlib.lines as mlines
import matplotlib.patches as mpatches
import matplotlib.ticker as ticker
import warnings
warnings.filterwarnings('ignore')

# ------------------------------------------------------------------------------
# 2. DATA LOADING & PREPROCESSING
# ------------------------------------------------------------------------------
# Load the raw dataset
file_path = '/content/drive/My Drive/EucVision/01. Master Dataset.csv'
df = pd.read_csv(file_path, low_memory=False)

# Clean, simple numerical date parsing
df['Date'] = pd.to_datetime(df['Date'], format='%d-%m-%Y', errors='coerce')

# Ensure critical measurement columns are treated as numeric
numeric_cols = [
    'Crown_Area',
    'Tree_Height',
    'Stem_Diameter',
    'Ground_Truth_Height',
    'LiDAR_Height',
    'Spacing'
]

for col in numeric_cols:
    df[col] = pd.to_numeric(df[col], errors='coerce')

# Generate a unique identifier for each tree to track growth over time
df['Tree_ID'] = (df['Compartment'].astype(str) + '_' +
                 df['Line'].astype(str) + '_' +
                 df['Plot'].astype(str) + '_' +
                 df['Tree'].map('{:.2f}'.format))

# Collapse split rows before math calculations happen
group_cols = ['Date', 'Compartment', 'Line', 'Plot', 'Culture', 'Spacing', 'Species', 'Tree', 'Tree_ID']

if 'Death_Date' in df.columns:
    group_cols.append('Death_Date')

df = df.groupby(group_cols, dropna=False).first().reset_index()

# Sort chronologically by tree to prepare for sequential calculations
df = df.sort_values(by=['Tree_ID', 'Date'])

# ------------------------------------------------------------------------------
# 2.5 MORTALITY FILTERING
# ------------------------------------------------------------------------------
if 'Death_Date' in df.columns:
    df['Parsed_Death_Date'] = pd.to_datetime(df['Death_Date'], format='%d-%m-%Y', errors='coerce')
    df['Is_Dead'] = df['Parsed_Death_Date'].notna() & (df['Date'] >= df['Parsed_Death_Date'])

    metrics_to_wipe = ['Crown_Area', 'Tree_Height', 'Stem_Diameter', 'Ground_Truth_Height', 'LiDAR_Height']
    df.loc[df['Is_Dead'], metrics_to_wipe] = np.nan
    df = df.drop(columns=['Parsed_Death_Date', 'Is_Dead'])
    print("Mortality Filter Applied: Dead tree metrics successfully wiped.")

# ------------------------------------------------------------------------------
# 3. HEIGHT ADJUSTMENT PIPELINE (SPECIES ARCHITECTURE + EPOCHS)
# ------------------------------------------------------------------------------

# Store original raw height for reference
df['Original_Height'] = df['Tree_Height']

# Biological constraint validation on raw height
df = df.sort_values(by=['Tree_ID', 'Date'])
df['Tree_Height'] = df.groupby('Tree_ID')['Tree_Height'].cummax()

# 1. EXTRACT GROUND TRUTHS
target_date_tls = pd.to_datetime('2025-11-28')
target_date_field = pd.to_datetime('2026-03-23')
target_date_als = pd.to_datetime('2026-06-26')
cutoff_date = pd.to_datetime('2026-03-16')   # TLS -> Field model switch
cutoff_date_als = target_date_als             # Field -> ALS model switch

tls_ref = df[df['Date'] == target_date_tls][['Tree_ID', 'LiDAR_Height']].rename(columns={'LiDAR_Height': 'Target_Height'})
field_ref = df[df['Date'] == target_date_field][['Tree_ID', 'Ground_Truth_Height']].rename(columns={'Ground_Truth_Height': 'Target_Height'})
als_field_ref = df[df['Date'] == target_date_als][['Tree_ID', 'Ground_Truth_Height']].rename(columns={'Ground_Truth_Height': 'Target_Height'})

df['Target_Height'] = np.nan

mask_tls = df['Date'] == target_date_tls
df.loc[mask_tls, 'Target_Height'] = df.loc[mask_tls, 'Tree_ID'].map(tls_ref.set_index('Tree_ID')['Target_Height'])

mask_field = df['Date'] == target_date_field
df.loc[mask_field, 'Target_Height'] = df.loc[mask_field, 'Tree_ID'].map(field_ref.set_index('Tree_ID')['Target_Height'])

mask_als = df['Date'] == target_date_als
df.loc[mask_als, 'Target_Height'] = df.loc[mask_als, 'Tree_ID'].map(als_field_ref.set_index('Tree_ID')['Target_Height'])
# 2. TRAIN CALIBRATION MODELS (Per Species, Per Epoch)
calibration_models = []

for target_date in [target_date_tls, target_date_field, target_date_als]:
    date_data = df[df['Date'] == target_date].dropna(subset=['Target_Height', 'Tree_Height', 'Species'])

    for species, group in date_data.groupby('Species'):
        if len(group) > 5:  # Minimum N to fit a stable line
            X = group[['Tree_Height']]
            y = group['Target_Height']
            reg = LinearRegression().fit(X, y)

            calibration_models.append({
                'Epoch_Date': target_date,
                'Species': species,
                'Slope': reg.coef_[0],
                'Intercept': reg.intercept_
            })
        else:
            # Fallback for insufficient species data: apply simple mean offset
            mean_diff = (group['Target_Height'] - group['Tree_Height']).mean() if len(group) > 0 else 0
            calibration_models.append({
                'Epoch_Date': target_date,
                'Species': species,
                'Slope': 1.0,
                'Intercept': mean_diff
            })

model_df = pd.DataFrame(calibration_models)

# 3. APPLY THE PREDICTIVE MODELS
df['Calibrated_Height'] = df['Tree_Height']

# Only apply UAV calibration to flights on or after Sept 1, 2025
uav_start_date = pd.to_datetime('2025-09-01')
is_uav = df['Date'] >= uav_start_date
df_uav = df[is_uav].copy()

mask_before_switch = df_uav['Date'] < cutoff_date
mask_mid_switch = (df_uav['Date'] >= cutoff_date) & (df_uav['Date'] < cutoff_date_als)
mask_after_als = df_uav['Date'] >= cutoff_date_als

# Apply November models (TLS) to pre-switch drone data
nov_models = model_df[model_df['Epoch_Date'] == target_date_tls].set_index('Species')
for species in df_uav['Species'].unique():
    if species in nov_models.index:
        m = nov_models.loc[species, 'Slope']
        c = nov_models.loc[species, 'Intercept']
        spc_mask = mask_before_switch & (df_uav['Species'] == species)
        df_uav.loc[spc_mask, 'Calibrated_Height'] = (df_uav.loc[spc_mask, 'Tree_Height'] * m) + c

# Apply March models (Field) to mid-window drone data
mar_models = model_df[model_df['Epoch_Date'] == target_date_field].set_index('Species')
for species in df_uav['Species'].unique():
    if species in mar_models.index:
        m = mar_models.loc[species, 'Slope']
        c = mar_models.loc[species, 'Intercept']
        spc_mask = mask_mid_switch & (df_uav['Species'] == species)
        df_uav.loc[spc_mask, 'Calibrated_Height'] = (df_uav.loc[spc_mask, 'Tree_Height'] * m) + c

# Apply June (ALS) models to post-June-26 drone data
als_models = model_df[model_df['Epoch_Date'] == target_date_als].set_index('Species')
for species in df_uav['Species'].unique():
    if species in als_models.index:
        m = als_models.loc[species, 'Slope']
        c = als_models.loc[species, 'Intercept']
        spc_mask = mask_after_als & (df_uav['Species'] == species)
        df_uav.loc[spc_mask, 'Calibrated_Height'] = (df_uav.loc[spc_mask, 'Tree_Height'] * m) + c

# Re-integrate and validate
df.loc[is_uav, 'Calibrated_Height'] = df_uav['Calibrated_Height']

# Final biological constraint validation to ensure no shrinking over time
df = df.sort_values(by=['Tree_ID', 'Date'])
df['Calibrated_Height'] = df.groupby('Tree_ID')['Calibrated_Height'].cummax()

# ------------------------------------------------------------------------------
# 3.5 STATISTICAL VALIDATION (K-FOLD OUT-OF-SAMPLE PREDICTIONS)
# ------------------------------------------------------------------------------
df['Honest_Calibrated_Height'] = df['Calibrated_Height']

for cal_date in [target_date_tls, target_date_field, target_date_als]:
    df_temp = df[df['Date'] == cal_date].copy()
    mask = df_temp['Target_Height'].notna() & df_temp['Tree_Height'].notna() & df_temp['Species'].notna()

    if mask.sum() > 20:
        # Create a dynamic matrix that mimics independent per-species regressions
        # by including Tree_Height, Species dummies, and their interaction terms
        X = pd.get_dummies(df_temp.loc[mask, ['Tree_Height', 'Species']], drop_first=True, dtype=float)

        interaction_cols = {}
        for col in X.columns:
            if col.startswith('Species_'):
                interaction_cols[f'Height_x_{col}'] = X['Tree_Height'] * X[col]
        X = pd.concat([X, pd.DataFrame(interaction_cols)], axis=1)

        y = df_temp.loc[mask, 'Target_Height']

        # 5-Fold cross-validation predicting actual target height directly
        cv = KFold(n_splits=5, shuffle=True, random_state=42)
        oos_preds = cross_val_predict(LinearRegression(), X, y, cv=cv)

        valid_indices = df_temp.loc[mask].index
        df.loc[valid_indices, 'Honest_Calibrated_Height'] = oos_preds

# Cleanup
df = df.drop(columns=['Target_Height'])

# ------------------------------------------------------------------------------
# 4. FORESTRY METRIC DERIVATION
# ------------------------------------------------------------------------------
df = df.sort_values('Date')

df['Crown_Diameter'] = 2 * np.sqrt(df['Crown_Area'] / np.pi)
df['Stem_Diameter'] = df['Stem_Diameter'] / 10
df['Basal_Area'] = np.pi * (df['Stem_Diameter'] / 2) ** 2

# ------------------------------------------------------------------------------
# 5. DATA AGGREGATION
# ------------------------------------------------------------------------------
agg_cols = [
    'Crown_Area', 'Crown_Diameter', 'Stem_Diameter', 'Basal_Area',
    'Ground_Truth_Height', 'LiDAR_Height', 'Original_Height',
    'Calibrated_Height', 'Honest_Calibrated_Height'
]

def filter_dominant_by_height(df_source, height_col='Calibrated_Height'):
    """Isolate the top-20% dominant trees by HEIGHT within each
    Date/Spacing/Species/Plot cell."""
    df_clean = df_source.dropna(
        subset=[height_col, 'Species', 'Spacing', 'Plot', 'Date']).copy()
    dominant = []
    for _, group in df_clean.groupby(['Date', 'Spacing', 'Species', 'Plot']):
        threshold = group[height_col].quantile(0.80)
        dominant.append(group[group[height_col] >= threshold])
    return pd.concat(dominant, ignore_index=True)

# Full-population daily means
daily_mean = df.groupby('Date')[agg_cols].mean()
daily_mean.columns = [f"{c}_mean" for c in daily_mean.columns]

# Dominant-cohort daily means (top 20% by height, per plot/species/spacing)
df_dominant = filter_dominant_by_height(df)
daily_dom = df_dominant.groupby('Date')[agg_cols].mean()
daily_dom.columns = [f"{c}_top_20_mean" for c in daily_dom.columns]

daily_agg = daily_mean.join(daily_dom, how='outer').reset_index()

# ------------------------------------------------------------------------------
# 6. PLOTTING CONFIGURATIONS
# ------------------------------------------------------------------------------
def plot_dynamic_panel_grid(date_str, custom_title, reference_label='TLS'):

    date_ts = pd.Timestamp(date_str)
    df_date = df[df['Date'] == date_ts].copy()

    if df_date.empty:
        print(f"No data for {custom_title}")
        return

    has_reference = 'LiDAR_Height' in df_date.columns and not df_date['LiDAR_Height'].isnull().all()
    has_field = 'Ground_Truth_Height' in df_date.columns and not df_date['Ground_Truth_Height'].isnull().all()

    if has_reference and has_field:
        plots = [
            ('LiDAR_Height', 'Original_Height', f'Reference {reference_label} height (m)', 'Uncalibrated UAV height (m)', f'Uncalibrated UAV vs reference {reference_label}', 'blue', False),
            ('LiDAR_Height', 'Honest_Calibrated_Height', f'Reference {reference_label} height (m)', 'Calibrated UAV height (m)', f'Calibrated UAV vs reference {reference_label}', 'orange', False),
            ('LiDAR_Height', 'Ground_Truth_Height', f'Reference {reference_label} height (m)', 'Field-measured height (m)', f'Field-measured vs reference {reference_label}', 'teal', True),
            ('Ground_Truth_Height', 'Honest_Calibrated_Height', 'Field-measured height (m)', 'Calibrated UAV height (m)', 'Calibrated UAV vs field-measured', 'purple', True)
        ]
    elif has_reference:
        plots = [
            ('LiDAR_Height', 'Original_Height', f'{reference_label} height (m)', 'Uncalibrated UAV height (m)', f'Uncalibrated UAV vs reference {reference_label}', 'blue', False),
            ('LiDAR_Height', 'Honest_Calibrated_Height', f'{reference_label} height (m)', 'Calibrated UAV height (m)', f'Calibrated UAV vs reference {reference_label}', 'orange', False)
        ]
    else:
        plots = [
            ('Ground_Truth_Height', 'Original_Height', 'Field-measured height (m)', 'Uncalibrated UAV height (m)', 'Uncalibrated UAV vs field-measured', 'green', True),
            ('Ground_Truth_Height', 'Honest_Calibrated_Height', 'Field-measured height (m)', 'Calibrated UAV height (m)', 'Calibrated UAV vs field-measured', 'purple', True)
        ]

    num_plots = len(plots)
    cols = 2
    rows = math.ceil(num_plots / cols)

    spacings_sorted = sorted(df['Spacing'].dropna().unique())
    custom_colors = ['#118AB2', '#EF476F', '#FFD166', '#06D6A0', '#073B4C', '#9D4EDD', '#F77F00']
    spacing_colors = {sp: custom_colors[i % len(custom_colors)] for i, sp in enumerate(spacings_sorted)}

    # --- CALCULATE GLOBAL AXIS LIMITS ---
    global_min = float('inf')
    global_max = float('-inf')

    for plot_info in plots:
        x_c, y_c = plot_info[0], plot_info[1]
        valid_data = df_date.dropna(subset=[x_c, y_c])
        if not valid_data.empty:
            c_min = min(valid_data[x_c].min(), valid_data[y_c].min())
            c_max = max(valid_data[x_c].max(), valid_data[y_c].max())
            if c_min < global_min: global_min = c_min
            if c_max > global_max: global_max = c_max

    # Apply 5% padding to the global limits
    if global_min != float('inf'):
        plot_min_val = global_min * 0.95
        plot_max_val = global_max * 1.05
    else:
        plot_min_val, plot_max_val = 0, 1

    global_lims = [plot_min_val, plot_max_val]
    # ------------------------------------

    # --- Figure height scales with rows, with fixed extra space for legends ---
    legend_space_inches = 0.5   # fixed inches reserved at bottom for shared legend
    panel_height_inches = FW / 2 * rows   # panels stay square-ish
    fig_height = panel_height_inches + legend_space_inches

    fig, axes = plt.subplots(rows, cols, figsize=(FW, fig_height))
    fig.suptitle(custom_title, y=0.98)

    axes_flat = axes.flatten() if rows > 1 else (axes if isinstance(axes, np.ndarray) else [axes])

    for i in range(len(axes_flat)):
        ax = axes_flat[i]
        if i < num_plots:
            x_col, y_col, x_lab, y_lab, subplot_title, fallback_color, use_spacing_colors = plots[i]
            sub = df_date.dropna(subset=[x_col, y_col, 'Spacing'])

            if len(sub) > 0:
                r2 = stats.linregress(sub[x_col], sub[y_col]).rvalue ** 2
                rmse = np.sqrt(mean_squared_error(sub[x_col], sub[y_col]))
                print(f"  ({chr(97+i)}) n = {len(sub)} | R2 = {r2:.2f} | RMSE = {rmse:.2f}")

                if use_spacing_colors:
                    for sp in spacings_sorted:
                        sp_data = sub[sub['Spacing'] == sp]
                        if not sp_data.empty:
                            ax.scatter(sp_data[x_col], sp_data[y_col],
                                       color=spacing_colors.get(sp, fallback_color),
                                       alpha=0.8, s=15, zorder=3, label=f"{int(sp)}x{int(sp)}m")
                else:
                    sns.scatterplot(x=x_col, y=y_col, data=sub,
                                    color=fallback_color, alpha=0.7, ax=ax, s=15, zorder=3)

                # Swap local variables for our newly calculated global ones
                min_val = plot_min_val
                max_val = plot_max_val
                lims = global_lims

                # 1:1 line — label only in first panel
                line_label = '1:1 line' if i == 0 else '_nolegend_'
                ax.plot(lims, lims, 'k--', linewidth=1.5, zorder=4, label=line_label)

                # Bias shading and annotations — first panel only
                if i == 0:
                    ax.fill_between(lims, min_val, lims, color='lightcoral', alpha=0.15, zorder=1)
                    ax.fill_between(lims, lims, max_val, color='lightblue', alpha=0.15, zorder=1)
                    bbox_props = dict(boxstyle="round,pad=0.2", fc="white", ec="none", alpha=0.8)
                    ax.text(max_val * 0.75, max_val * 0.25, 'Height\nunderestimated',
                            color='firebrick', weight='bold', ha='center', va='center',
                            zorder=5, bbox=bbox_props)
                    ax.text(max_val * 0.25, max_val * 0.55, 'Height\noverestimated',
                            color='steelblue', weight='bold', ha='center', va='center',
                            zorder=5, bbox=bbox_props)

                ax.set_xlim(lims)
                ax.set_ylim(lims)

                ax.xaxis.set_major_locator(ticker.MultipleLocator(1))
                ax.yaxis.set_major_locator(ticker.MultipleLocator(1))

                ax.set_title(f"{subplot_title}\n$R^2$ = {r2:.2f} | RMSE = {rmse:.2f}m")
                ax.set_xlabel(x_lab, weight='bold')
                ax.set_ylabel(y_lab, weight='bold')
                ax.set_aspect('equal', adjustable='box')
                ax.grid(False)

            else:
                ax.text(0.5, 0.5, 'Insufficient data pairs', ha='center', va='center')
                ax.axis('off')
        else:
            ax.axis('off')

    add_panel_labels(axes_flat, num_plots)

    # --- Legend block ---
    # 1:1 line legend — top-left panel only
    line_1to1 = mlines.Line2D([], [], color='k', linestyle='--', linewidth=1.5)
    axes_flat[0].legend(handles=[line_1to1], labels=['1:1 line'],
                        loc='upper center')

    # Shared spacing legend — collect handles from all panels
    handles_dict = {}
    for ax in axes_flat:
        h_arr, l_arr = ax.get_legend_handles_labels()
        for h, l in zip(h_arr, l_arr):
            if not l.startswith('_') and l != '1:1 line' and l not in handles_dict:
                handles_dict[l] = h

    if handles_dict:
        blank = mpatches.Rectangle((0, 0), 0, 0, fill=False, edgecolor='none', visible=False)
        handles_list = [blank] + list(handles_dict.values())
        labels_list  = ['Spacing:'] + list(handles_dict.keys())

        # bottom_fraction = legend space as a fraction of total figure height
        bottom_fraction = legend_space_inches / fig_height

        fig.legend(handles_list, labels_list,
                   loc='lower center',
                   ncol=len(labels_list),
                   bbox_to_anchor=(0.5, 0.0),      # anchor to very bottom of figure
                   bbox_transform=fig.transFigure,  # use figure coordinates, not axes
                   handletextpad=0.3,
                   columnspacing=1.2)

        # Reserve exactly legend_space_inches at the bottom, leave top for suptitle
        top_fraction = 1.0 - (0.35 / fig_height)
        plt.tight_layout(rect=[0, bottom_fraction, 1, top_fraction])
        fig.subplots_adjust(hspace=0.5)

    else:
        # No spacing legend needed — simpler layout
        top_fraction = 1.0 - (0.35 / fig_height)
        plt.tight_layout(rect=[0, 0.02, 1, top_fraction])

    clean_date = date_ts.strftime('%Y-%m-%d')
    plt.savefig(f'/content/drive/My Drive/EucVision/Figures/Correlations_{clean_date}.png')
    plt.show()

def plot_growth_trajectory(df, smooth_frac=0.3):

    # Aggregate mean heights per date
    agg_df = df.groupby('Date').agg(
        Mean_Original=('Original_Height', 'mean'),
        Mean_Adjusted=('Calibrated_Height', 'mean'),
        Mean_LiDAR=('LiDAR_Height', 'mean'),
        Mean_Field=('Ground_Truth_Height', 'mean')
    ).reset_index()

    fig, ax = plt.subplots(figsize=(FW, HW))

    uav_start_date = pd.to_datetime('2025-09-01')
    uav_res_split_date = pd.to_datetime('2026-03-16')

    # Split the timeline
    df_pre_uav = agg_df[agg_df['Date'] < uav_start_date].copy()
    df_uav = agg_df[agg_df['Date'] >= uav_start_date].copy()

    data_pre_uav = df_pre_uav.dropna(subset=['Mean_Original']).copy()

    # Connect pre-UAV line to the first UAV point for visual continuity
    if not data_pre_uav.empty and not df_uav.empty:
        first_uav_point = df_uav.dropna(subset=['Mean_Adjusted']).iloc[[0]].copy()
        first_uav_point['Mean_Original'] = first_uav_point['Mean_Adjusted'] # Anchor to the calibrated point
        data_pre_uav_connected = pd.concat([data_pre_uav, first_uav_point]).sort_values(by='Date')
    else:
        data_pre_uav_connected = data_pre_uav

    # --- 1. Field-measured (Pre-UAV) ---
    field_label_used = False
    if not data_pre_uav_connected.empty:
        ax.plot(data_pre_uav_connected['Date'], data_pre_uav_connected['Mean_Original'],
                color='red', linestyle='-', linewidth=1.5, label='_nolegend_')

    if not data_pre_uav.empty:
        ax.plot(data_pre_uav['Date'], data_pre_uav['Mean_Original'],
                color='red', linestyle='', marker='*', markersize=6, zorder=5, label='Field-measured')
        field_label_used = True

    # Plot any late-stage Field data (like 23 Mar 2026) as stars
    data_gt = agg_df.dropna(subset=['Mean_Field'])
    if not data_gt.empty:
        gt_label = '_nolegend_' if field_label_used else 'Field-measured'
        ax.plot(data_gt['Date'], data_gt['Mean_Field'],
                color='red', linestyle='', marker='*', markersize=6, zorder=5, label=gt_label)

    # --- 2. Uncalibrated UAV ---
    if not df_uav.empty:
        uav_orig_0_6cm = df_uav[df_uav['Date'] < uav_res_split_date].copy()
        uav_orig_3cm = df_uav[df_uav['Date'] >= uav_res_split_date].copy()

        # Connect 0.6cm to 3cm for continuity
        if not uav_orig_0_6cm.empty and not uav_orig_3cm.empty:
            first_3cm_point = uav_orig_3cm.dropna(subset=['Mean_Original']).iloc[[0]].copy()
            uav_orig_0_6cm_connected = pd.concat([uav_orig_0_6cm, first_3cm_point]).sort_values(by='Date')
        else:
            uav_orig_0_6cm_connected = uav_orig_0_6cm

        if not uav_orig_0_6cm_connected.empty:
            ax.plot(uav_orig_0_6cm_connected['Date'], uav_orig_0_6cm_connected['Mean_Original'],
                    color='gray', linestyle=':', linewidth=1.5, marker='o', markersize=4, label='Uncalibrated 0.6cm GSD')

        if not uav_orig_3cm.empty:
            ax.plot(uav_orig_3cm['Date'], uav_orig_3cm['Mean_Original'],
                    color='darkblue', linestyle=':', linewidth=1.5, marker='o', markersize=4, label='Uncalibrated 3cm GSD')

    # --- 3. Calibrated UAV (LOESS Smooth) ---
    color_ht_raw = "#66CC66"
    color_ht_smooth = "#008000"

    data_uav_adj = df_uav.dropna(subset=['Mean_Adjusted']).copy()
    if not data_uav_adj.empty:
        # Plot raw calibrated points slightly faded
        ax.plot(data_uav_adj['Date'], data_uav_adj['Mean_Adjusted'],
                color=color_ht_raw, linestyle='-', linewidth=1.0, marker='o', markersize=4, alpha=0.5, label='_nolegend_')

        # Apply LOESS
        clean_ht = data_uav_adj[['Date', 'Mean_Adjusted']].dropna()
        if len(clean_ht) > 3:
            x_num_ht = mdates.date2num(clean_ht['Date'])
            y_val_ht = clean_ht['Mean_Adjusted'].values
            smoothed_ht = lowess(y_val_ht, x_num_ht, frac=smooth_frac)
            ax.plot(clean_ht['Date'], smoothed_ht[:, 1],
                    color=color_ht_smooth, linewidth=2, label='Calibrated UAV')

# --- 4. Reference LiDAR (TLS vs ALS) ---
    data_lidar = agg_df.dropna(subset=['Mean_LiDAR'])
    data_lidar_tls = data_lidar[data_lidar['Date'] < target_date_als]
    data_lidar_als = data_lidar[data_lidar['Date'] >= target_date_als]

    if not data_lidar_tls.empty:
        ax.plot(data_lidar_tls['Date'], data_lidar_tls['Mean_LiDAR'],
                color='lightblue', markeredgecolor='#0066cc', linestyle='', marker='D', markersize=6, zorder=4, label='Reference TLS')

    if not data_lidar_als.empty:
        ax.plot(data_lidar_als['Date'], data_lidar_als['Mean_LiDAR'],
                color='#FFD166', markeredgecolor='#F77F00', linestyle='', marker='D', markersize=6, zorder=1.5, label='Reference ALS')

    # --- Formatting ---
    ax.set_title("Effect of calibration on stand height growth trajectory", pad=10)
    ax.set_ylabel("Mean height per tree (m)")

    ax.set_ylim(bottom=0)

    ax.xaxis.set_major_locator(mdates.MonthLocator(interval=2))
    ax.xaxis.set_major_formatter(mdates.DateFormatter('%b %Y'))
    ax.tick_params(axis='x', rotation=45)
    ax.tick_params(axis='y')

    # Extract all line labels dynamically for the trajectory plot
    h_arr, l_arr = ax.get_legend_handles_labels()
    handles_dict = {}
    for h, l in zip(h_arr, l_arr):
        if not l.startswith('_') and l not in handles_dict:
            handles_dict[l] = h

    # Match the trajectory layout to the panel grid layout
    fig.legend(handles_dict.values(), handles_dict.keys(),
               loc='lower center',
               ncol=3,  # Adjusted to 3 so the 5 trajectory lines fit well
               bbox_to_anchor=(0.5, -0.06))

    plt.tight_layout(rect=[0, 0.06, 1, 0.94])

    # Save the trajectory dashboard
    plt.savefig('/content/drive/My Drive/EucVision/Figures/Stand_Height_Growth_Trajectory.png')
    plt.show()

# ------------------------------------------------------------------------------
# 6.5 DATA CLEANING: MERGE SPLIT ROWS AND SORT
# ------------------------------------------------------------------------------
group_cols = ['Date', 'Compartment', 'Line', 'Plot', 'Culture', 'Spacing', 'Species', 'Tree', 'Tree_ID']
if 'Death_Date' in df.columns: group_cols.append('Death_Date')

df = df.groupby(group_cols, dropna=False).first().reset_index()
df = df.sort_values(by=['Date', 'Plot', 'Tree'], ascending=[True, True, True])

# Run Visualizations
plot_dynamic_panel_grid('2025-11-28', 'Tree height correlations: 28 November 2025 (0.6cm GSD)', reference_label='TLS')
plot_dynamic_panel_grid('2026-03-23', 'Tree height correlations: 23 March 2026 (3cm GSD)', reference_label='TLS')
plot_dynamic_panel_grid('2026-06-26', 'Tree height correlations: 26 June 2026 (3cm GSD)', reference_label='ALS')
plot_growth_trajectory(df)

# ------------------------------------------------------------------------------
# 7. EXPORT PROCESSED DATA
# ------------------------------------------------------------------------------
column_order = [
    'Date', 'Compartment', 'Line', 'Plot', 'Culture', 'Spacing', 'Species', 'Tree','Tree_ID',
    'Death_Date','Stem_Diameter', 'Crown_Area', 'Calibrated_Height', #'Honest_Calibrated_Height',
    'LiDAR_Height','Ground_Truth_Height', 'Canopy_Cover', 'Crown_Diameter', 'Basal_Area'
]

unit_mapping = {
    'Stem_Diameter': 'Stem_Diameter_cm',
    'Crown_Area': 'Crown_Area_m2',
    'Calibrated_Height': 'Calibrated_Height_m',
    #'Honest_Calibrated_Height': 'Honest_Calibrated_Height_m',
    'LiDAR_Height': 'LiDAR_Height_m',
    'Ground_Truth_Height': 'Field_Measured_Height_m',
    'Crown_Diameter': 'Crown_Diameter_m',
    'Canopy_Cover': 'Crown_Cover_Index',
    'Basal_Area': 'Basal_Area_cm2',
}

export_cols = [c for c in column_order if c in df.columns]
export_df = df[export_cols].rename(columns=unit_mapping)

# Extract the latest flight date and format it as YYYY-MM-DD
latest_date_str = df['Date'].max().strftime('%d-%m-%Y')

# Inject the date string into the file paths
main_output_path = f'/content/drive/My Drive/EucVision/UAV_Master_Dataset_{latest_date_str}.csv'
export_df.to_csv(main_output_path, index=False)
print(f"Master dataset successfully saved to: {main_output_path}")

# Ensure Honest validation columns are dropped from aggregate if they snuck in
cols_to_drop = [c for c in daily_agg.columns if 'Honest' in c]
daily_agg = daily_agg.drop(columns=cols_to_drop, errors='ignore')

agg_output_path = f'/content/drive/My Drive/EucVision/UAV_Aggregated_Metrics_{latest_date_str}.csv'
daily_agg.to_csv(agg_output_path, index=False)
print(f"Aggregated metrics successfully saved to: {agg_output_path}")
```

```python
# @title Height Accuracy Summary Table

# ------------------------------------------------------------------------------
# A. SOURCE REGISTRY
# ------------------------------------------------------------------------------
# rank sets the X-axis convention: lower rank always becomes the reference.
SOURCES = {
    'LiDAR':    dict(col='LiDAR_Height',             rank=0, kind='ref'),
    'Field':    dict(col='Ground_Truth_Height',      rank=1, kind='ref'),
    'UAV_raw':  dict(col='Original_Height',          rank=2, kind='uav'),
    'UAV_mono': dict(col='Tree_Height',              rank=3, kind='uav'),
    'UAV_cal':  dict(col='Honest_Calibrated_Height', rank=4, kind='uav'),
}

LABELS = {'Field':    'Field-measured',
          'UAV_raw':  'Uncalibrated UAV (raw)',
          'UAV_mono': 'Uncalibrated UAV (monotonic)',
          'UAV_cal':  'Calibrated UAV'}

# Which reference each calibration epoch was actually trained against
CAL_TARGET = {target_date_tls: 'LiDAR', target_date_field: 'Field', target_date_als: 'Field'}
GSD_SWITCH = pd.Timestamp('2026-03-16')

def lidar_label(d):
    return 'ALS' if d >= target_date_als else 'TLS'

def validation_label(d, ref_key, cmp_key):
    if SOURCES[cmp_key]['kind'] == 'ref':
        return 'Reference vs reference'
    if cmp_key != 'UAV_cal':
        return 'Uncalibrated'
    if d in CAL_TARGET:
        return ('Cross-validated (training target)' if CAL_TARGET[d] == ref_key
                else 'Cross-validated (independent reference)')
    return 'Model applied (no CV)'

# ------------------------------------------------------------------------------
# B. METRICS
# ------------------------------------------------------------------------------
def pair_metrics(x, y):
    x = np.asarray(x, dtype=float)
    y = np.asarray(y, dtype=float)
    lr = stats.linregress(x, y)
    resid = y - x
    rmse = np.sqrt(np.mean(resid ** 2))
    return dict(N=len(x),
                R2_OLS=lr.rvalue ** 2,        # symmetric: strength of association
                R2_1to1=r2_score(x, y),       # asymmetric: agreement about the 1:1 line
                Slope=lr.slope,
                Intercept=lr.intercept,
                RMSE=rmse,
                rRMSE_pct=100 * rmse / np.mean(x),
                MAE=np.mean(np.abs(resid)),
                Bias=np.mean(resid))          # positive = comparison overestimates

# ------------------------------------------------------------------------------
# C. BUILD EVERY AVAILABLE PAIR, EVERY DATE
# ------------------------------------------------------------------------------
MIN_N = 5
rows = []

for date, d in df.groupby('Date'):
    present = {k: v for k, v in SOURCES.items()
               if v['col'] in d.columns and d[v['col']].notna().sum() >= MIN_N}
    keys = sorted(present, key=lambda k: present[k]['rank'])

    for i, ref_key in enumerate(keys):
        for cmp_key in keys[i + 1:]:
            # skip UAV-vs-UAV: that measures the size of our own adjustment, not accuracy
            if present[ref_key]['kind'] == 'uav' and present[cmp_key]['kind'] == 'uav':
                continue

            sub = d[[present[ref_key]['col'], present[cmp_key]['col']]].dropna()
            if len(sub) < MIN_N:
                continue

            rows.append(dict(
                Date=date,
                GSD='3 cm' if date >= GSD_SWITCH else '0.6 cm',
                Reference=lidar_label(date) if ref_key == 'LiDAR' else LABELS[ref_key],
                Comparison=lidar_label(date) if cmp_key == 'LiDAR' else LABELS[cmp_key],
                Validation=validation_label(date, ref_key, cmp_key),
                **pair_metrics(sub.iloc[:, 0], sub.iloc[:, 1])))

long_df = pd.DataFrame(rows)

# ------------------------------------------------------------------------------
# D. EXPORT: TIDY LONG + PUBLICATION-SHAPED
# ------------------------------------------------------------------------------
stamp = df['Date'].max().strftime('%d-%m-%Y')
base = '/content/drive/My Drive/EucVision'

long_df.to_csv(f'{base}/Height_Accuracy_Metrics_Long_{stamp}.csv', index=False)

pub = long_df.copy()
pub['Comparison'] = pub['Comparison'] + ' vs ' + pub['Reference']
pub['Date'] = pub['Date'].dt.strftime('%d %b %Y')

fmt = {'R2_OLS': '{:.2f}', 'R2_1to1': '{:.2f}', 'Slope': '{:.2f}', 'Intercept': '{:+.2f}',
       'RMSE': '{:.2f}', 'rRMSE_pct': '{:.1f}', 'MAE': '{:.2f}', 'Bias': '{:+.2f}'}
for c, f in fmt.items():
    pub[c] = pub[c].map(f.format)

pub = pub[['Date', 'GSD', 'Comparison', 'Validation', 'N',
           'R2_OLS', 'R2_1to1', 'Slope', 'Intercept',
           'RMSE', 'rRMSE_pct', 'MAE', 'Bias']]
pub.columns = ['Date', 'GSD', 'Comparison', 'Validation', 'n',
               'R\u00b2 (OLS)', 'R\u00b2 (1:1)', 'Slope', 'Intercept',
               'RMSE (m)', 'rRMSE (%)', 'MAE (m)', 'Bias (m)']

# blank repeated Date/GSD so the Word table reads as grouped blocks
pub.loc[pub['Date'].duplicated(), ['Date', 'GSD']] = ''

pub.to_csv(f'{base}/Height_Accuracy_Metrics_Table_{stamp}.csv',
           index=False, encoding='utf-8-sig')

print(f"Long-format metrics: {len(long_df)} comparisons across "
      f"{long_df['Date'].nunique()} dates")
display(pub)
```

```python
# @title Heights Compared
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
import scipy.stats as stats
import matplotlib.ticker as ticker

# 1. Define Google Colab File Paths
base_path = '/content/drive/My Drive/EucVision/'
file_1 = base_path + '03. 28 November 2025 Heights Compared.csv'
file_2 = base_path + '04. 02 March 2026 Densifications Compared.csv'
file_3 = base_path + '05. 23 March 2026 Heights Compared.csv'

# Load Data
df1 = pd.read_csv(file_1)
df2 = pd.read_csv(file_2)
df3 = pd.read_csv(file_3)

# Strip any whitespace from column names
df1.columns = df1.columns.str.strip()
df2.columns = df2.columns.str.strip()
df3.columns = df3.columns.str.strip()

# ------------------------------------------------------------------------------
# CONSISTENCY FIX: use cross-validated calibrated heights, matching Table 3.1
# ------------------------------------------------------------------------------
def prefer_cross_validated(df, name):
    """Swap in the out-of-sample calibrated height so every calibrated figure
    in the thesis reports the same value. Renames to the original column name
    so all downstream labels and column lists are unaffected."""
    if 'Honest_Calibrated_Height_m' in df.columns:
        df = df.drop(columns=['Calibrated_Height_m'], errors='ignore')
        df = df.rename(columns={'Honest_Calibrated_Height_m': 'Calibrated_Height_m'})
        print(f"{name}: using cross-validated calibrated height.")
    else:
        print(f"{name}: WARNING — Honest_Calibrated_Height_m not found. "
              f"Falling back to IN-SAMPLE Calibrated_Height_m; these values will "
              f"NOT match Table 3.1.")
    return df

df1 = prefer_cross_validated(df1, 'File 1 (28 Nov 2025)')
df2 = prefer_cross_validated(df2, 'File 2 (02 Mar 2026)')
df3 = prefer_cross_validated(df3, 'File 3 (23 Mar 2026)')

# Collector for Table 3.2
table_32_rows = []

# Helper function to calculate R^2, filter, and plot to avoid repetition
def plot_height_distribution(df, ref_col, title, cols_to_compare=None, compartment_filter=None,
                             custom_labels=None, block_label=None, reference_name=None):
    df_copy = df.copy()

    # Filter for compartment if specified
    if compartment_filter is not None:
        df_copy = df_copy[df_copy['Compartment'] == compartment_filter]

    # Gather all columns to process
    if cols_to_compare is None:
        cols_to_compare = [c for c in df_copy.columns if 'Height' in c or '_m' in c]

    all_cols = [ref_col] + [c for c in cols_to_compare if c != ref_col and c in df_copy.columns]

    # Filter out extreme anomalies (<0m and >10m)
    for col in all_cols:
        if col in df_copy.columns:
            mask = (df_copy[col] < 0) | (df_copy[col] > 10)
            df_copy.loc[mask, col] = np.nan

    # Calculate R^2 (OLS) and error metrics against the reference column
    r2_dict = {}
    print(f"\n--- {title} ---")
    for col in all_cols:
        if col not in df_copy.columns:
            continue

        valid_data = df_copy[[ref_col, col]].dropna()

        if col == ref_col or len(valid_data) < 2:
            r2_dict[col] = 1.0
            continue

        lr    = stats.linregress(valid_data[ref_col], valid_data[col])
        resid = valid_data[col] - valid_data[ref_col]
        rmse  = np.sqrt((resid ** 2).mean())
        r2_dict[col] = lr.rvalue ** 2

        # Plain-text method name for the table
        method = custom_labels.get(col, col) if custom_labels else col
        method = method.replace('\n', ' ').strip()

        table_32_rows.append(dict(
            Block=block_label or title,
            Reference=reference_name or ref_col,
            Method=method,
            N=len(valid_data),
            R2_OLS=lr.rvalue ** 2,
            Slope=lr.slope,
            Intercept=lr.intercept,
            RMSE=rmse,
            rRMSE_pct=100 * rmse / valid_data[ref_col].mean(),
            MAE=resid.abs().mean(),
            Bias=resid.mean()))

        print(f"  {method}: n = {len(valid_data)} | R2 = {lr.rvalue ** 2:.2f} | "
              f"slope = {lr.slope:.2f} | RMSE = {rmse:.2f} | "
              f"MAE = {resid.abs().mean():.2f} | bias = {resid.mean():+.2f}")

    plt.figure(figsize=(FW, HW))
    df_melted = df_copy.melt(value_vars=[c for c in all_cols if c in df_copy.columns],
                             var_name='Method', value_name='Height (m)')

    # Formatting labels to include custom names and R^2
    labels = {}
    for col in all_cols:
        if col in df_copy.columns:
            base_name = col
            if custom_labels and col in custom_labels:
                base_name = custom_labels[col]

            if col == ref_col:
                labels[col] = f"{base_name}\n(Reference)"
            elif col in r2_dict:
                labels[col] = f"{base_name}\n$R^2$: {r2_dict[col]:.2f}"

    df_melted['Method_Label'] = df_melted['Method'].map(labels)

    sns.violinplot(
        x='Method_Label',
        y='Height (m)',
        hue='Method_Label',
        data=df_melted,
        inner="quartile",
        palette="Set2",
        legend=False
    )

    ax = plt.gca()
    ax.set_ylim(bottom=0)
    ax.yaxis.set_major_locator(ticker.MultipleLocator(1))

    plt.title(title, pad=15)
    plt.ylabel('Tree height (m)')
    plt.xlabel('')
    plt.tight_layout()

    safe_title = title.replace(" ", "_").replace(":", "").replace("(", "").replace(")", "")
    output_path = f"/content/drive/My Drive/EucVision/Figures/{safe_title}.png"
    plt.savefig(output_path)
    plt.show()


# ==========================================
# FILE 1: 28 November 2025
# ==========================================
november_labels = {
    'LiDAR_Height_m' : 'TLS',
    'Ground_Truth_Height_m': 'Field-measured',
    'Calibrated_Height_m': 'Calibrated UAV',
    'DTM_Height_m': 'DTM derived',
    'CSF_Height_m': 'CSF derived',
    'DSM_Height_m': 'DSM derived'
}

plot_height_distribution(
    df=df1,
    ref_col='LiDAR_Height_m',
    cols_to_compare=['Ground_Truth_Height_m','Calibrated_Height_m', 'DTM_Height_m', 'CSF_Height_m', 'DSM_Height_m'],
    title='Tree height distribution - 28 Nov 2025',
    compartment_filter=None,
    custom_labels=november_labels,
    block_label='28 November 2025 (0.6 cm GSD)',
    reference_name='TLS'
)

# ==========================================
# FILE 2: 02 March 2026
# ==========================================
densification_labels = {
    'Ground_Truth_Height_m': 'Field-measured',
    '4.8_Double_m': '4.8 cm/pt\nCross hatch',
    '4.8_Single_m': '4.8 cm/pt\nSingle',
    '2.4_Single_m': '2.4 cm/pt\nSingle',
    '19.2_Single_m': '19.2 cm/pt\nSingle',
    'Calibrated_Height_m': 'Calibrated UAV'
}

plot_height_distribution(
    df=df2,
    ref_col='Ground_Truth_Height_m',
    cols_to_compare=['Calibrated_Height_m', '4.8_Double_m', '4.8_Single_m', '2.4_Single_m', '19.2_Single_m'],
    title='Tree height distribution - 02 March 2026 (compartment 2)',
    compartment_filter=2,
    custom_labels=densification_labels,
    block_label='2 March 2026, Compartment 2 (0.6 cm GSD)',
    reference_name='Field-measured'
)

# ==========================================
# FILE 3: 23 March 2026
# ==========================================
march_labels = {
    'LiDAR_Height_m' : 'TLS',
    'Ground_Truth_Height_m': 'Field-measured',
    'Calibrated_Height_m': 'Calibrated UAV',
    'DTM_3cm_Height_m': 'DTM derived\n [3cm GSD]',
    'DTM_0.6cm_Height_m': 'DTM derived\n[0.6cm GSD]',
    'CSF_Height_m': 'CSF derived\n[3cm GSD]',
    'DSM_Height_m': 'DSM derived\n[3cm GSD]'
}

plot_height_distribution(
    df=df3,
    ref_col='Ground_Truth_Height_m',
    cols_to_compare=['Calibrated_Height_m','DTM_3cm_Height_m', 'CSF_Height_m', 'DTM_0.6cm_Height_m', 'DSM_Height_m'],
    title='Tree height distribution - 23 March 2026',
    compartment_filter=None,
    custom_labels=march_labels,
    block_label='23 March 2026 (3 cm GSD)',
    reference_name='Field-measured'
)

# ------------------------------------------------------------------------------
# TABLE 3.2 EXPORT
# ------------------------------------------------------------------------------
long_32 = pd.DataFrame(table_32_rows)
long_32.to_csv(f'{base_path}Height_Extraction_Method_Metrics_Long.csv', index=False)

pub_32 = long_32.copy()
pub_32['Comparison'] = pub_32['Method'] + ' vs ' + pub_32['Reference']

fmt = {'R2_OLS': '{:.2f}', 'Slope': '{:.2f}', 'Intercept': '{:+.2f}',
       'RMSE': '{:.2f}', 'rRMSE_pct': '{:.1f}', 'MAE': '{:.2f}', 'Bias': '{:+.2f}'}
for c, f in fmt.items():
    pub_32[c] = pub_32[c].map(f.format)

pub_32 = pub_32[['Block', 'Comparison', 'N', 'R2_OLS', 'Slope',
                 'RMSE', 'rRMSE_pct', 'MAE', 'Bias']]
pub_32.columns = ['Date', 'Comparison', 'n', 'R\u00b2 (OLS)', 'Slope',
                  'RMSE (m)', 'rRMSE (%)', 'MAE (m)', 'Bias (m)']

# blank repeated date so the Word table reads as grouped blocks
pub_32.loc[pub_32['Date'].duplicated(), 'Date'] = ''

pub_32.to_csv(f'{base_path}Height_Extraction_Method_Table.csv',
              index=False, encoding='utf-8-sig')

print(f"\nTable 3.2: {len(long_32)} method comparisons exported.")
display(pub_32)
```

```python
# @title Flight Report
import os
import io
import datetime
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.dates as mdates
import matplotlib.lines as mlines
from matplotlib.patches import Patch
from matplotlib.transforms import offset_copy

# =====================================================================
# 1. DIRECTORY SETUP & FILE LOADING
# =====================================================================
FIGDIR = '/content/drive/My Drive/EucVision/Figures'
os.makedirs(FIGDIR, exist_ok=True)

REPORT_DIR = '/content/drive/My Drive/EucVision/Flight Reports and Canopy Coverage/'
flight_file = REPORT_DIR + '07. Flight Reports Summary.csv'
planar_file = REPORT_DIR + '06. Planar Coverage and Point Density.csv'

# --- A. Load and Clean Main Flight Data (07) ---
df_flights = pd.read_csv(flight_file)

df_flights['pct_calibrated'] = df_flights['% images calibrated'].str.rstrip('%').astype(float)
df_flights['match_ratio_pct'] = (df_flights['Median matched keypoints per image'] / df_flights['Median keypoints per image']) * 100
df_flights['Date'] = pd.to_datetime(df_flights['Date'], format='mixed', dayfirst=True)

# FILTER: Remove 'Normal' flights for April 29th across all graphs
is_apr29 = (df_flights['Date'].dt.month == 4) & (df_flights['Date'].dt.day == 29)
is_normal = df_flights['Flight path'].str.contains('Normal', case=False, na=False)
df_flights = df_flights[~(is_apr29 & is_normal)]

# Start the whole report from the 30 Oct 2025 flight (drops the 1 Sept 2025 outlier flight)
EARLIEST_FLIGHT = pd.Timestamp('2025-10-30')
df_flights = df_flights[df_flights['Date'] >= EARLIEST_FLIGHT]

df_flights['Area_ha'] = df_flights['Area Covered'].astype(str).str.extract(r'([\d.]+)\s*ha').astype(float)
df_flights['Area_m2'] = df_flights['Area_ha'] * 10000
df_flights['point_density_m2'] = df_flights['Total 3D Densified points'] / df_flights['Area_m2']

# Calculate Photos per Hectare globally for the Photos/ha row
df_flights['Total Images'] = pd.to_numeric(df_flights['Total Images'], errors='coerce')
df_flights['Photos_per_ha'] = df_flights['Total Images'] / df_flights['Area_ha'].replace(0, np.nan)

def classify_regime(row):
    gsd = row['GSD (cm)']
    if row['Flight path'] == 'Cross Hatch': return 'Cross Hatch (~3cm GSD)'
    elif gsd <= 0.68: return 'Normal & Cross (\u22640.70cm GSD)'
    elif gsd <= 1.2: return 'Normal (0.70-1.00cm GSD)'

df_flights['regime'] = df_flights.apply(classify_regime, axis=1)

def get_marker(row):
    if row.get('Compartment') == 'Top': return 'o'
    if row.get('Compartment') == 'Bottom': return 's'
df_flights['marker'] = df_flights.apply(get_marker, axis=1)

regime_colors = {
    'Normal & Cross (\u22640.70cm GSD)': '#0072B2',
    'Normal (0.70-1.00cm GSD)': '#009E73',
    'Cross Hatch (~3cm GSD)': '#7B2CBF',
}
regime_order = [r for r in regime_colors.keys() if r in df_flights['regime'].unique()]

# --- B. Load Planar Coverage & Density Data (06) ---
df_planar = pd.read_csv(planar_file)
df_planar['Flight_Date'] = pd.to_datetime(df_planar['Flight_Date'], dayfirst=True)

# Remove April 29th from planar coverage data
is_planar_apr29 = (df_planar['Flight_Date'].dt.month == 4) & (df_planar['Flight_Date'].dt.day == 29)
df_planar = df_planar[~is_planar_apr29]
df_planar = df_planar[df_planar['Flight_Date'] >= EARLIEST_FLIGHT]

if not df_planar.empty:
    df_planar['Plot_Num'] = df_planar['Plot_ID'].str.extract(r'(\d+)').astype(int)
    df_planar['Compartment_P'] = np.where(df_planar['Plot_Num'] <= 21, 'Top', 'Bottom')

    # --- Plot -> Spacing lookup (used to colour the new spacing coverage panel) ---
    # NOTE: `spacing_colors` itself is assumed to already exist as a global variable
    # earlier in the notebook (same convention as MW/FW/HW/TW), so it is not redefined here.
    _plot_spacing_raw = """Space\tPlot_P
1\tP1
5\tP2
2\tP3
3\tP4
2\tP5
1\tP6
3\tP7
2\tP8
1\tP9
3\tP10
5\tP11
2\tP12
5\tP13
2\tP14
3\tP15
1\tP16
3\tP17
1\tP18
5\tP19
3\tP20
2\tP21
1\tP22
3\tP23
3\tP24
1\tP25
5\tP26
3\tP27
2\tP28
5\tP29
1\tP30
2\tP31
2\tP32
3\tP33
1\tP34
5\tP35
3\tP36
1\tP37
2\tP38
3\tP39
5\tP40
5\tP41
3\tP42
2\tP43
1\tP44
5\tP45
2\tP46
3\tP47
1\tP48
3\tP49
3\tP50
2\tP51
5\tP52
3\tP53
1\tP54
2\tP55
5\tP56
1\tP57
3\tP58
1\tP59
3\tP60
2\tP61
5\tP62
3\tP63
1\tP64
2\tP65
3\tP66
3\tP67
5\tP68
3\tP69
1\tP70
5\tP71
2\tP72
1\tP73
2\tP74
3\tP75"""
    _spacing_lookup = pd.read_csv(io.StringIO(_plot_spacing_raw), sep='\t')
    _spacing_lookup['Plot_Num'] = _spacing_lookup['Plot_P'].str.extract(r'(\d+)').astype(int)
    plot_spacing_map = dict(zip(_spacing_lookup['Plot_Num'], _spacing_lookup['Space']))
    df_planar['Spacing'] = df_planar['Plot_Num'].map(plot_spacing_map)

    planar_agg = df_planar.groupby(['Flight_Date', 'Compartment_P']).agg({
        'Coverage_Percent': 'mean',
        'Canopy_Density_3cm': 'mean',
        'Ground_Density_3cm': 'mean',
        'Canopy_Density_06cm': 'mean',
        'Ground_Density_06cm': 'mean',
    }).reset_index().sort_values(['Flight_Date', 'Compartment_P'])
else:
    planar_agg = pd.DataFrame(columns=['Flight_Date', 'Compartment_P', 'Coverage_Percent', 'Canopy_Density_3cm',
                                        'Ground_Density_3cm', 'Canopy_Density_06cm', 'Ground_Density_06cm'])

# Merge planar aggregates onto the flight table
df_flights = pd.merge(df_flights, planar_agg, left_on=['Date', 'Compartment'], right_on=['Flight_Date', 'Compartment_P'], how='left')

# Canopy Matched Keypoint Percentage
is_hatch = df_flights['regime'] == 'Cross Hatch (~3cm GSD)'
df_flights['Density_Canopy'] = np.where(is_hatch, df_flights['Canopy_Density_3cm'], df_flights['Canopy_Density_06cm'])
df_flights['Density_Ground'] = np.where(is_hatch, df_flights['Ground_Density_3cm'], df_flights['Ground_Density_06cm'])

safe_ground_density = df_flights['Density_Ground'].replace(0, np.nan)
df_flights['canopy_match_pct'] = df_flights['match_ratio_pct'] / safe_ground_density * df_flights['Density_Canopy']

# --- C. Load Daily Weather / Wind Data ---
path_new_24 = '/content/drive/My Drive/EucVision/impact_NDVI_2024-01-01_2024-12-31.xlsx'
path_new_25 = '/content/drive/My Drive/EucVision/impact_NDVI_2025-01-01_2025-12-31.xlsx'
path_new_26 = '/content/drive/My Drive/EucVision/impact_NDVI_2026-01-01_2026-05-31.xlsx'

try:
    df_weather = pd.concat([pd.read_excel(path_new_24), pd.read_excel(path_new_25), pd.read_excel(path_new_26)], ignore_index=True)
    df_weather['Date'] = pd.to_datetime(df_weather['Date'], errors='coerce')
    df_weather = df_weather.dropna(subset=['Date']).sort_values('Date').reset_index(drop=True)
    df_weather['Wind Speed (m/s)'] = pd.to_numeric(df_weather['Wind Speed (m/s)'], errors='coerce')
except FileNotFoundError:
    print("Warning: Weather files not found. Daily wind plot will be empty.")
    df_weather = pd.DataFrame(columns=['Date', 'Wind Speed (m/s)'])

# =====================================================================
# 2. SHARED STYLE, HELPERS & DATE BOUNDS
# =====================================================================
date_fmt = mdates.DateFormatter('%b %Y')

STYLE = dict(s=70, alpha=0.88, edge='#1a1a1a', lw=0.9)
MARKER_PIXEL_OFFSET = {'o': -3.5, 's': 3.5, '^': 0.0, 'd': 0.0}
TOP_COLOR = '#E69F00'
BOTTOM_COLOR = '#56B4E9'
spacing_labels = {
    1: '1\u00d71m',
    2: '2\u00d72m',
    3: '3\u00d73m',
    5: '5\u00d75m',
}
COVERAGE_START = pd.Timestamp('2026-03-01')

fancy_legend_kwargs = {
    'loc': 'upper center',
    'bbox_to_anchor': (0.5, 1.06),
    'ncol': 2
}

if not df_planar.empty:
    unique_dates = pd.to_datetime(sorted(df_planar['Flight_Date'].unique()))
    date_nums = mdates.date2num(unique_dates.to_pydatetime())
    box_width = max(np.median(np.diff(date_nums)) * 0.5, 1.5) if len(date_nums) > 1 else 3
else:
    unique_dates, date_nums, box_width = [], np.array([]), 3

if not df_flights.empty:
    _full_dates = pd.to_datetime(sorted(set(df_flights['Date'].unique()) | set(df_planar['Flight_Date'].unique())))
    _full_date_nums = mdates.date2num(_full_dates.to_pydatetime())
    _full_pad = max(np.median(np.diff(_full_date_nums)) * 0.5, 1.5) if len(_full_date_nums) > 1 else 3
    shared_xmin = _full_date_nums.min() - _full_pad * 2.5
    shared_xmax = _full_date_nums.max() + _full_pad * 2.5
else:
    shared_xmin, shared_xmax = None, None

def set_limits_from_data(ax_target, y_vals, y_pad_frac=0.08):
    y_vals = pd.Series(y_vals).dropna()
    if not y_vals.empty:
        y_min, y_max = y_vals.min(), y_vals.max()
        y_pad = (y_max - y_min) * y_pad_frac or 1
        ax_target.set_ylim(y_min - y_pad, y_max + y_pad)

df_coverage = df_planar[df_planar['Flight_Date'] >= COVERAGE_START] if not df_planar.empty else df_planar
if not df_coverage.empty:
    cov_unique_dates = pd.to_datetime(sorted(df_coverage['Flight_Date'].unique()))
    cov_date_nums = mdates.date2num(cov_unique_dates.to_pydatetime())
    cov_box_width = max(np.median(np.diff(cov_date_nums)) * 0.5, 1.5) if len(cov_date_nums) > 1 else 3
    cov_xmin = cov_date_nums.min() - cov_box_width * 2.5
else:
    cov_unique_dates, cov_date_nums, cov_box_width = [], np.array([]), 3
    cov_xmin = mdates.date2num(COVERAGE_START)
cov_xmax = shared_xmax if shared_xmax is not None else (cov_date_nums.max() + cov_box_width * 2.5 if len(cov_date_nums) else None)

def plot_complex_scatter(fig, ax_target, x_col, y_col):
    for regime in regime_order:
        for marker in df_flights['marker'].unique():
            sub = df_flights[(df_flights['regime'] == regime) & (df_flights['marker'] == marker)]
            if len(sub) == 0 or sub[y_col].isna().all(): continue
            dx = MARKER_PIXEL_OFFSET.get(marker, 0.0)
            trans = offset_copy(ax_target.transData, fig=fig, x=dx, y=0, units='points')
            ax_target.scatter(
                sub[x_col], sub[y_col], transform=trans,
                color=regime_colors[regime], marker=marker,
                s=STYLE['s'], alpha=STYLE['alpha'],
                edgecolor=STYLE['edge'], linewidth=STYLE['lw'], zorder=3
            )
    set_limits_from_data(ax_target, df_flights[y_col])

def plot_grouped_coverage(ax_target, source_df, group_col, group_specs, dates, dnums, box_w, seed=42):
    if source_df.empty:
        return
    rng = np.random.default_rng(seed)
    for d, dnum in zip(dates, dnums):
        sub = source_df[source_df['Flight_Date'] == d]
        for gval, marker, color, z_ord in group_specs:
            gsub = sub[sub[group_col] == gval]
            if gsub.empty: continue
            jitter = rng.uniform(-box_w * 0.32, box_w * 0.32, size=len(gsub))
            ax_target.scatter(
                dnum + jitter, gsub['Coverage_Percent'],
                marker=marker, color=color, s=30, alpha=0.85,
                edgecolor='#1a1a1a', linewidth=0.5, zorder=z_ord
            )

def build_master_legend(fig, ax_target):
    legend_elements = []
    for regime in regime_order:
        legend_elements.append(Patch(facecolor=regime_colors[regime], edgecolor='#1a1a1a', linewidth=0.8, label=regime))
    legend_elements.append(mlines.Line2D([], [], color='none', marker='o', markerfacecolor='#d9d9d9', markeredgecolor='#1a1a1a', markersize=7, label='Top compartment'))
    legend_elements.append(mlines.Line2D([], [], color='none', marker='s', markerfacecolor='#d9d9d9', markeredgecolor='#1a1a1a', markersize=7, label='Bottom compartment'))
    ax_target.legend(handles=legend_elements, loc='upper left', bbox_to_anchor=(0.08, 1.0))

# =====================================================================
# 3. GRAPH 1 — 2x1: Wind & Flight Dates / GCP RMS Error
# =====================================================================
fig1, (axWind, axRMSE) = plt.subplots(2, 1, figsize=(MW, 4), sharex=True)
plt.subplots_adjust(hspace=0.15)

if not df_weather.empty:
    axWind.plot(df_weather['Date'], df_weather['Wind Speed (m/s)'], color='teal', linewidth=0.8, label='Wind speed', zorder=3)
    flight_wind = df_weather[df_weather['Date'].isin(df_flights['Date'].dt.floor('D'))]
    axWind.scatter(flight_wind['Date'], flight_wind['Wind Speed (m/s)'],
                    color='orange', edgecolor='darkgoldenrod',
                    s=25, linewidths=0.5, zorder=4, label='UAV flight dates')
    axWind.legend(**fancy_legend_kwargs)
else:
    axWind.text(0.5, 0.5, "No Weather Data Available", va='center', ha='center', transform=axWind.transAxes)
axWind.set_ylabel('Wind speed\n(m/s)')
axWind.set_ylim(top=10)

if not df_flights.empty:
    plot_complex_scatter(fig1, axRMSE, 'Date', 'GCP RMS error (m)')
    axRMSE.set_ylabel('GCP RMSE\n(m)')
else:
    axRMSE.axis('off')
    axRMSE.text(0.5, 0.5, "No Flight Data Available", va='center', ha='center', transform=axRMSE.transAxes)

if shared_xmin and shared_xmax:
    axRMSE.set_xlim(shared_xmin, shared_xmax)
axRMSE.xaxis_date()
axRMSE.xaxis.set_major_formatter(date_fmt)
axRMSE.tick_params(axis='x', rotation=30)

add_panel_labels([axWind, axRMSE], 2)

build_master_legend(fig1, axRMSE)
plt.suptitle('UAV flights: wind conditions and GCP accuracy', y=0.97)
save_path_1 = os.path.join(FIGDIR, '2_Row_Flight_Report.png')
fig1.savefig(save_path_1, dpi=300, bbox_inches='tight')
plt.show()

# =====================================================================
# 4. GRAPH 2 — 3 rows, bottom row split: Photos/ha | Canopy Matched % | Coverage (Spacing | Compartment)
# =====================================================================
fig2 = plt.figure(figsize=(MW, 7))

gs = fig2.add_gridspec(2, 1, height_ratios=[2, 1.15], hspace=0.30, top=0.92)
gs_top = gs[0].subgridspec(2, 1, hspace=0.08)
gs_bot = gs[1].subgridspec(1, 2, wspace=0.15)

axPhotos2 = fig2.add_subplot(gs_top[0])
axKeypts = fig2.add_subplot(gs_top[1], sharex=axPhotos2)
axCovSpacing = fig2.add_subplot(gs_bot[0])
axCovComp = fig2.add_subplot(gs_bot[1], sharey=axCovSpacing)

if not df_flights.empty:
    plot_complex_scatter(fig2, axPhotos2, 'Date', 'Photos_per_ha')
    axPhotos2.set_ylabel('Mean photos per hectare')
else:
    axPhotos2.axis('off')
    axPhotos2.text(0.5, 0.5, "No Flight Data Available", va='center', ha='center', transform=axPhotos2.transAxes)
plt.setp(axPhotos2.get_xticklabels(), visible=False)

plot_complex_scatter(fig2, axKeypts, 'Date', 'canopy_match_pct')
axKeypts.set_ylabel('Canopy keypoints\nmatched (%)')
axKeypts.xaxis_date()
axKeypts.xaxis.set_major_formatter(date_fmt)
axKeypts.tick_params(axis='x', rotation=30, labelbottom=True)

spacing_order = [1, 2, 3, 5]
df_coverage_bottom = df_coverage[df_coverage['Compartment_P'] == 'Bottom'] if not df_coverage.empty else df_coverage
if not df_coverage_bottom.empty:
    box_data = [df_coverage_bottom.loc[df_coverage_bottom['Spacing'] == sp, 'Coverage_Percent'].dropna().values for sp in spacing_order]
    bp = axCovSpacing.boxplot(
        box_data, positions=range(len(spacing_order)), widths=0.55,
        patch_artist=True, showfliers=False, zorder=3,
        medianprops=dict(color='#1a1a1a', linewidth=1.4),
        whiskerprops=dict(color='#1a1a1a', linewidth=1.0),
        capprops=dict(color='#1a1a1a', linewidth=1.0),
        boxprops=dict(edgecolor='#1a1a1a', linewidth=1.0),
    )
    for patch, sp in zip(bp['boxes'], spacing_order):
        patch.set_facecolor(spacing_colors[sp])
        patch.set_alpha(0.55)

    _rng_sp = np.random.default_rng(42)
    for i, sp in enumerate(spacing_order):
        vals = df_coverage_bottom.loc[df_coverage_bottom['Spacing'] == sp, 'Coverage_Percent'].dropna()
        if vals.empty: continue
        jitter = _rng_sp.uniform(-0.18, 0.18, size=len(vals))
        axCovSpacing.scatter(i + jitter, vals, color=spacing_colors[sp], s=20, alpha=0.75,
                              edgecolor='#1a1a1a', linewidth=0.4, zorder=4)

axCovSpacing.set_xlim(-0.6, len(spacing_order) - 0.4)
axCovSpacing.set_xticks(range(len(spacing_order)))
axCovSpacing.set_xticklabels([spacing_labels[sp].split(' (')[0] for sp in spacing_order])
axCovSpacing.set_ylabel('Bottom plots\ncanopy coverage (%)')

compartment_specs = [('Bottom', 's', BOTTOM_COLOR, 3), ('Top', 'o', TOP_COLOR, 4)]
plot_grouped_coverage(axCovComp, df_coverage, 'Compartment_P', compartment_specs,
                      cov_unique_dates, cov_date_nums, cov_box_width)
plt.setp(axCovComp.get_yticklabels(), visible=False)
axCovComp.set_ylabel('Canopy\ncoverage (%)')
top_handle = mlines.Line2D([], [], color='none', marker='o', markerfacecolor=TOP_COLOR, markeredgecolor='#1a1a1a', markersize=7, label='Top plots')
bottom_handle = mlines.Line2D([], [], color='none', marker='s', markerfacecolor=BOTTOM_COLOR, markeredgecolor='#1a1a1a', markersize=7, label='Bottom plots')
axCovComp.legend(handles=[top_handle, bottom_handle], loc='lower left', ncol=1, fontsize=7)

if not df_coverage.empty:
    set_limits_from_data(axCovSpacing, df_coverage['Coverage_Percent'])

if cov_xmin is not None and cov_xmax is not None:
    axCovComp.set_xlim(cov_xmin, cov_xmax)
axCovComp.xaxis_date()
axCovComp.xaxis.set_major_locator(mdates.MonthLocator())
axCovComp.xaxis.set_major_formatter(date_fmt)
axCovComp.tick_params(axis='x', rotation=30)

if shared_xmin and shared_xmax:
    axPhotos2.set_xlim(shared_xmin, shared_xmax)

pos_top = axKeypts.get_position()
pos_bot = axCovSpacing.get_position()
fig2.text(0.5, 0.39, 'Sub-1cm GSD canopy coverage (~3cm GSD reference)',
          ha='center', va='center', fontsize=11, fontweight='bold')

add_panel_labels([axPhotos2, axKeypts, axCovSpacing, axCovComp], 4)

build_master_legend(fig2, axKeypts)
plt.suptitle('UAV flights: imaging quality and canopy coverage', y=0.96)
save_path_2 = os.path.join(FIGDIR, '3_Row_Flight_Report.png')
fig2.savefig(save_path_2, dpi=300, bbox_inches='tight')
plt.show()
```

```python
# @title Environmental Drivers

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.dates as mdates
import matplotlib.patches as mpatches

# =====================================================================
# 1. LOAD & CLEAN DATA
# =====================================================================
global_start_date = pd.to_datetime('2025-09-01')

# --- A. Weather Data (Impact) ---
path_new_25 = '/content/drive/My Drive/EucVision/impact_NDVI_2025-01-01_2025-12-31.xlsx'
path_new_26 = '/content/drive/My Drive/EucVision/impact_NDVI_2026-01-01_2026-05-31.xlsx'

df_25 = pd.read_excel(path_new_25)
df_26 = pd.read_excel(path_new_26)
df_weather = pd.concat([df_25, df_26], ignore_index=True)

df_weather['Date'] = pd.to_datetime(df_weather['Date'], errors='coerce')
df_weather = df_weather.dropna(subset=['Date']).sort_values('Date').reset_index(drop=True)
df_weather = df_weather[df_weather['Date'] >= global_start_date]

# Auto-detect VPD columns if they exist in the spreadsheet
vpd_cols = [c for c in df_weather.columns if 'VPD' in c.upper()]
cols_to_num = ['Max Deg.C', 'Min Deg.C', 'Precipitation (mm)', 'Wind Speed (m/s)', 'Humidity (%)'] + vpd_cols

for c in cols_to_num:
    df_weather[c] = pd.to_numeric(df_weather[c], errors='coerce')

# If VPD columns aren't directly in the data, calculate them using Temp & RH
if len(vpd_cols) < 2:
    es_max = 0.6108 * np.exp((17.27 * df_weather['Max Deg.C']) / (df_weather['Max Deg.C'] + 237.3))
    es_min = 0.6108 * np.exp((17.27 * df_weather['Min Deg.C']) / (df_weather['Min Deg.C'] + 237.3))
    ea = (es_max + es_min) / 2 * (df_weather['Humidity (%)'] / 100)
    df_weather['Max_VPD_Calc'] = es_max - ea
    df_weather['Min_VPD_Calc'] = es_min - ea
    vpd_max_col, vpd_min_col = 'Max_VPD_Calc', 'Min_VPD_Calc'
else:
    vpd_max_col = next((c for c in vpd_cols if 'MAX' in c.upper()), vpd_cols[0])
    vpd_min_col = next((c for c in vpd_cols if 'MIN' in c.upper()), vpd_cols[1])

weekly_data = df_weather.resample('W-WED', on='Date')['Precipitation (mm)'].sum().reset_index()
weekly_data.rename(columns={'Precipitation (mm)': 'Weekly_Precip'}, inplace=True)
weekly_data['Plot_Date'] = weekly_data['Date'] - pd.Timedelta(days=3.5)

# --- B. Gems Data (Main Updated) ---
# 1. Load the original March data
path_gems = '/content/drive/My Drive/EucVision/Main_Gems_March_2026.csv'
df_gems_march = pd.read_csv(path_gems, low_memory=False)

# 2. Load the new April and May data
path_gems_apr = '/content/drive/My Drive/EucVision/GEMS - April 2026.csv'
path_gems_may = '/content/drive/My Drive/EucVision/GEMS - May 2026.csv'

df_gems_apr = pd.read_csv(path_gems_apr, low_memory=False)
df_gems_may = pd.read_csv(path_gems_may, low_memory=False)

# Combine new data
df_gems_new = pd.concat([df_gems_apr, df_gems_may], ignore_index=True)

# 3. Rename new columns to match the old dataset's structure
rename_map = {
    'Time': 'Datetime',
    'Apogee Pyro.Solar.1.4': 'solar_radiation',
    'Acclima Soil.Temperature.1.1': 'soil_temp_north',
    'Acclima Soil.Temperature.1.2': 'soil_temp_south',
    'Acclima Soil.VWC.1.1': 'soil_vwc_north',
    'Acclima Soil.VWC.1.2': 'soil_vwc_south'
}
df_gems_new.rename(columns=rename_map, inplace=True)

# 4. Concatenate all GEMS data together
df_gems = pd.concat([df_gems_march, df_gems_new], ignore_index=True)

# 5. Proceed with original cleaning and processing
df_gems['Datetime'] = pd.to_datetime(df_gems['Datetime'], errors='coerce')
df_gems = df_gems.dropna(subset=['Datetime'])

df_gems = df_gems[df_gems['Datetime'] >= global_start_date].copy()
df_gems['DateOnly'] = df_gems['Datetime'].dt.floor('D')

df_gems['solar_radiation'] = pd.to_numeric(df_gems['solar_radiation'], errors='coerce').clip(lower=0)

for col in ['soil_temp_north', 'soil_temp_south']:
    s = pd.to_numeric(df_gems[col], errors='coerce')
    df_gems[col] = s.where((s > -20) & (s < 60))

for col in ['soil_vwc_north', 'soil_vwc_south']:
    s = pd.to_numeric(df_gems[col], errors='coerce')
    df_gems[col] = s.where((s >= 0) & (s <= 100))

daily_gems = df_gems.groupby('DateOnly').agg({
    'solar_radiation': 'mean',
    'soil_temp_north': ['max', 'min'],
    'soil_temp_south': ['max', 'min'],
    'soil_vwc_north': ['max', 'min']
})
daily_gems.columns = ['_'.join(col) for col in daily_gems.columns]
daily_gems = daily_gems.reset_index()

# --- ADDED: Weekly Cumulative Energy Calculation ---
# 1 W/m² = 1 Joule/second. Multiply by seconds in a day (86400), divide by 1,000,000 for Megajoules
daily_gems['daily_radiation_MJ'] = (daily_gems['solar_radiation_mean'] * 86400) / 1000000

weekly_rad = daily_gems.resample('W-WED', on='DateOnly')['daily_radiation_MJ'].sum().reset_index()
weekly_rad.rename(columns={'daily_radiation_MJ': 'Weekly_Solar_Rad'}, inplace=True)
weekly_rad['Plot_Date'] = weekly_rad['DateOnly'] - pd.Timedelta(days=3.5)
# ---------------------------------------------------

# =====================================================================
# 2. HELPER SETUP & 3x2 GRID ALIGNMENT
# =====================================================================

fig, axs = plt.subplots(3, 2, figsize=(MW, 6), sharex=True)

# Increased hspace to 0.25 so lifted legends don't hit the x-axis above them
plt.subplots_adjust(hspace=0.25, wspace=0.25)
fig.suptitle("Environmental conditions during the UAV survey period (Sep 2025 – May 2026)", y=0.96)

# Base legend parameters with lifted position
fancy_legend_kwargs = {
    'loc': 'upper center',        # Aligns the bottom-center of the legend box
    'bbox_to_anchor': (0.5,1.14) # Places it at x=0.5 (middle) and y=1.14 (just above top spine)
}

# =====================================================================
# 3. COLUMN 1: Atmospheric Drivers (Above-Ground)
# =====================================================================

# --- Panel 1 (0,0): Solar Radiation (UPDATED) ---
axs[0, 0].plot(weekly_rad['Plot_Date'], weekly_rad['Weekly_Solar_Rad'], color='#d35400', linewidth=0.8, zorder=3, label='Weekly cumulative solar energy')
axs[0, 0].fill_between(weekly_rad['Plot_Date'], 0, weekly_rad['Weekly_Solar_Rad'], color='#d35400', alpha=0.2, zorder=2)
axs[0, 0].set_ylabel('Weekly cumulative\nsolar energy (MJ/m²)')
axs[0, 0].legend(**fancy_legend_kwargs)

# --- Panel 2 (1,0): Vapor Pressure Deficit (SWAPPED) ---
axs[1, 0].plot(df_weather['Date'], df_weather[vpd_max_col], color='#8e44ad', linewidth=0.8, label='Max vapor pressure deficit')
axs[1, 0].set_ylabel('Daily vapor pressure deficit\n(kPa)')
axs[1, 0].legend(**fancy_legend_kwargs)

# --- Panel 3 (2,0): Air Temp (SWAPPED) ---
axs[2, 0].plot(df_weather['Date'], df_weather['Max Deg.C'], color='#d62728', linewidth=0.8, label='Max air temp')
axs[2, 0].plot(df_weather['Date'], df_weather['Min Deg.C'], color='#1f77b4', linewidth=0.8, label='Min air temp')
axs[2, 0].fill_between(df_weather['Date'], df_weather['Min Deg.C'], df_weather['Max Deg.C'], color='grey', alpha=0.15)
axs[2, 0].set_ylabel('Daily air temperature\n(°C)')
axs[2, 0].legend(ncol=2, **fancy_legend_kwargs) # Force 2 columns


# =====================================================================
# 4. COLUMN 2: Soil & Hydrological Drivers (Below-Ground)
# =====================================================================

# --- Panel 4 (0,1): Weekly Precip ---
axs[0, 1].bar(weekly_data['Plot_Date'], weekly_data['Weekly_Precip'], color='#4B8BBE', edgecolor='black', linewidth=0.5, width=6, zorder=3)
axs[0, 1].set_ylabel('Weekly rainfall\n(mm)')

precip_patch = mpatches.Patch(color='#4B8BBE', label='Weekly rainfall')
apr_patch = mpatches.Patch(color='orange', alpha=0.3, label="April'26 storm")
may_patch = mpatches.Patch(color='red', alpha=0.2, label="May'26 storm")

axs[0, 1].legend(handles=[precip_patch, apr_patch, may_patch], **fancy_legend_kwargs)

# --- Panel 5 (1,1): Soil Moisture (North VWC Only) ---
mean_vwc = daily_gems[['soil_vwc_north_max', 'soil_vwc_north_min']].mean(axis=1)
axs[1, 1].plot(daily_gems['DateOnly'], mean_vwc, color='#27ae60', linewidth=0.8, label='Mean soil moisture')
axs[1, 1].fill_between(daily_gems['DateOnly'], daily_gems['soil_vwc_north_min'], daily_gems['soil_vwc_north_max'], color='#2ecc71', alpha=0.3)
axs[1, 1].set_ylabel('Hourly soil moisture\n(%)')
axs[1, 1].legend(**fancy_legend_kwargs)

# --- Panel 6 (2,1): Soil Temp (South Only) ---
axs[2, 1].plot(daily_gems['DateOnly'], daily_gems['soil_temp_south_max'], color='#d62728', linewidth=0.8, label='Max soil temp')
axs[2, 1].plot(daily_gems['DateOnly'], daily_gems['soil_temp_south_min'], color='#1f77b4', linewidth=0.8, label='Min soil temp')
axs[2, 1].fill_between(daily_gems['DateOnly'], daily_gems['soil_temp_south_min'], daily_gems['soil_temp_south_max'], color='grey', alpha=0.15)
axs[2, 1].set_ylabel('Daily soil temperature\n(°C)')
axs[2, 1].legend(ncol=2, **fancy_legend_kwargs) # Force 2 columns


# =====================================================================
# 5. GLOBAL FORMATTING & ALIGNMENT
# =====================================================================
date_apr = pd.to_datetime('2026-04-19')
date_may = pd.to_datetime('2026-05-11')

for ax in axs.flat:
    # Storm Background Bars
    ax.axvspan(date_apr - pd.Timedelta(days=4), date_apr + pd.Timedelta(days=3), color='orange', alpha=0.3, zorder=2)
    ax.axvspan(date_may - pd.Timedelta(days=5), date_may + pd.Timedelta(days=2), color='red', alpha=0.2, zorder=2)

# Y-Axis Limits & Synchronization
axs[0, 1].set_ylim(0,175) # Precip
axs[0, 0].set_ylim(0, weekly_rad['Weekly_Solar_Rad'].max() * 1.35) # Solar Energy
axs[1, 0].set_ylim(0,7) # Vapor Pressure Deficit
axs[1, 1].set_ylim(bottom=0) # Soil Moisture

# Lock Air Temp (now axs[2,0]) and Soil Temp (axs[2,1]) to the exact same Y-axis scale starting at 0
temp_max = max(df_weather['Max Deg.C'].max(), daily_gems['soil_temp_south_max'].max()) + 5
axs[2, 0].set_ylim(0, temp_max) # Air Temp
axs[2, 1].set_ylim(0, temp_max) # Soil Temp

# X-Axis Formatting for the bottom row panels
max_date = max(df_weather['Date'].max(), daily_gems['DateOnly'].max())
for ax in [axs[2, 0], axs[2, 1]]:
    # Changed interval to 2 to prevent overlapping labels on the smaller figure width
    ax.xaxis.set_major_locator(mdates.MonthLocator(interval=2))
    ax.xaxis.set_major_formatter(mdates.DateFormatter('%b %Y'))
    ax.tick_params(axis='x', rotation=45)
    ax.set_xlim(global_start_date, max_date)

add_panel_labels([axs[0, 0], axs[0, 1], axs[1, 0], axs[1, 1],axs[2, 0],axs[2, 1]], 6)

# Export the figure using the global template's dpi and bbox settings
plt.savefig('/content/drive/My Drive/EucVision/Figures/3x2_Environmental_Conditions.png')
plt.show()
```

```python
# @title Tree Metrics Overview

import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.dates as mdates
import numpy as np
from pygam import LinearGAM, s

# ==============================================================================
# HELPER: Fit GAM with Dynamic Splines and Lambda Penalization
# ==============================================================================
def fit_true_gam(dates_series, values_series, n_splines, lam=0.6):
    """
    Fits a GAM to raw individual data.
    Accepts 'lam' (Lambda) to penalize flexibility and prevent sagging in data gaps.
    """
    x_num = mdates.date2num(pd.to_datetime(dates_series).values)
    y = values_series.values

    unique_dates = len(np.unique(x_num))

    if unique_dates > 3:
        X = x_num.reshape(-1, 1)
        actual_splines = min(n_splines, unique_dates)

        try:
            # Apply the lambda penalty to stiffen the curve through empty gaps!
            gam = LinearGAM(s(0, n_splines=actual_splines, lam=lam)).fit(X, y)

            x_smooth = np.linspace(x_num.min(), x_num.max(), 100)
            X_smooth = x_smooth.reshape(-1, 1)
            y_pred = gam.predict(X_smooth)

            # +/- 1 Standard Error based on individual tree variance (68.27% CI)
            y_intervals = gam.confidence_intervals(X_smooth, width=0.6827)

            return mdates.num2date(x_smooth), y_pred, y_intervals
        except Exception:
            pass # Fallback to raw if singular matrix

    return None, None, None


# ==============================================================================
# MASTER PLOTTING SCRIPT
# ==============================================================================

# Custom tuning per metric! Crown area gets stiffened (lam=50) to cross the 7-month gap safely.
metrics = [
    {'col': 'Crown_Area',       'ylabel': 'Mean crown area\n per tree (m²)',  'splines': 15,  'lam': 50.0},
    {'col': 'Calibrated_Height', 'ylabel': 'Mean calibrated height\n per tree (m)',       'splines': 15, 'lam': 50.0},
    {'col': 'Basal_Area',      'ylabel': 'Mean basal area\n per tree (cm²)', 'splines': 12, 'lam': 0.6}
]

# (Ensure FW is defined globally in your notebook)
fig, axes = plt.subplots(3, 1, figsize=(FW, FW), sharex=True)
fig.suptitle("Individual-tree metrics by spacing", fontweight='bold')

spacings_asc = sorted(df['Spacing'].dropna().unique())
spacings_sorted = sorted(spacings_asc, reverse=True)

custom_colors = ['#118AB2', '#EF476F', '#FFD166', '#06D6A0', '#073B4C', '#9D4EDD', '#F77F00']
spacing_colors = {sp: custom_colors[i % len(custom_colors)] for i, sp in enumerate(spacings_asc)}

planting_date = pd.to_datetime('2024-05-01')
global_min_date = df['Date'].min() - pd.Timedelta(days=15)
global_max_date = df['Date'].max() + pd.Timedelta(days=15)

for idx, metric_info in enumerate(metrics):
    ax = axes[idx]
    metric_col = metric_info['col']

    # Using the exact column names from your raw CSV
    df_clean = df.dropna(subset=[metric_col, 'Spacing', 'Date']).copy()
    df_clean['Date'] = pd.to_datetime(df_clean['Date'])

    # --------------------------------------------------------------------------
    # ALL TREES — mean per spacing per date
    # --------------------------------------------------------------------------
    df_sp_raw = df_clean.copy()

    for spacing_val in spacings_sorted:
        sp_data = df_sp_raw[df_sp_raw['Spacing'] == spacing_val]
        if sp_data.empty: continue

        label = f"{int(spacing_val)}x{int(spacing_val)}m"
        color = spacing_colors.get(int(spacing_val), '#808080')

        # Pass the metric-specific splines and lambda parameters
        x_dates, mean_gam, se_gam = fit_true_gam(sp_data['Date'], sp_data[metric_col],
                                                 n_splines=metric_info['splines'],
                                                 lam=metric_info['lam'])

        if x_dates is not None:
            ax.plot(x_dates, mean_gam, color=color, linewidth=2.0, label=label, zorder=4)
            ax.fill_between(x_dates, se_gam[:, 0], se_gam[:, 1], color=color, alpha=0.15, zorder=3)

            # Get the actual mean for each exact date you flew/collected data
            actual_means = sp_data.groupby('Date')[metric_col].mean().reset_index()
            # Overlay those actual dates as points behind the GAM line
            ax.scatter(actual_means['Date'], actual_means[metric_col],
                       color=color, s=15, alpha=0.4, zorder=3.5)

        else:
            # Fallback if GAM fails
            fallback_agg = sp_data.groupby('Date')[metric_col].agg(['mean', 'std', 'count']).reset_index()
            fallback_agg['se'] = fallback_agg['std'] / np.sqrt(fallback_agg['count'])
            ax.errorbar(fallback_agg['Date'], fallback_agg['mean'], yerr=fallback_agg['se'],
                        color=color, linewidth=2.0, marker='o', markersize=4, label=label, zorder=4)

    # --------------------------------------------------------------------------
    # AXIS FORMATTING & AESTHETICS
    # --------------------------------------------------------------------------
    ax.set_ylabel(metric_info['ylabel'], fontweight='bold')
    ax.set_ylim(bottom=0)
    ax.set_xlim(left=global_min_date, right=global_max_date)
    ax.tick_params(axis='y')

    if idx == 2:
        ax.xaxis.set_major_locator(mdates.MonthLocator(interval=2))
        ax.xaxis.set_major_formatter(mdates.DateFormatter('%b %Y'))
        ax.tick_params(axis='x', rotation=45)
    ax.set_xlabel('')

# --------------------------------------------------------------------------
# AGE MARKERS (Purple Vertical Lines)
# --------------------------------------------------------------------------
for i, ax in enumerate(axes):
    intervals_in_months = [12, 24, 36, 48, 60]
    for months in intervals_in_months:
        v_date = planting_date + pd.DateOffset(months=months)
        if v_date > global_max_date: break

        ax.axvline(x=v_date, color='purple', linestyle='--', linewidth=1.5, alpha=0.8, zorder=1)

        # Draw the age text only on the bottom plot to keep it clean
        if i == 2:
            ymin, ymax = ax.get_ylim()
            y_span = ymax - ymin
            y_pos = ymin + (y_span * 0.85)

            text_x_pos = v_date - pd.Timedelta(days=5)

            ax.text(text_x_pos, y_pos, f'Age: {months//12}',
                    color='purple', fontweight='bold',
                    verticalalignment='center', horizontalalignment='right')

# --------------------------------------------------------------------------
# FINAL RENDER
# --------------------------------------------------------------------------
handles, labels = axes[0].get_legend_handles_labels()
by_label = dict(zip(labels, handles))

axes[0].legend(by_label.values(), by_label.keys(), loc='center left')

add_panel_labels([axes[0],axes[1],axes[2]], 3)

plt.tight_layout(rect=[0, 0, 1, 1], h_pad=1.5)
plt.savefig('/content/drive/My Drive/EucVision/Figures/Stand_Overview.png')
plt.show()
```

```python
# @title Absolute Growth

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.dates as mdates
from pygam import LinearGAM, s

# Calculate Conical Stem Volume and Crown Volume
df['Stem_Volume'] = (1/3) * (df['Basal_Area'] / 10000) * df['Calibrated_Height']
df['Crown_Volume'] = (1/3) * df['Crown_Area'] * df['Calibrated_Height']

# ==============================================================================
# HELPER: Extract Top 20% Individuals (Strictly by Height!)
# ==============================================================================
def filter_dominant_by_height(df_source, height_col='Calibrated_Height'):
    """
    For each Date/Spacing/Species/Plot, isolates the top-20% dominant trees
    based strictly on HEIGHT.
    """
    # Ensure our height column exists to do the filtering
    df_clean = df_source.dropna(subset=[height_col, 'Species', 'Spacing', 'Plot', 'Date']).copy()
    df_clean['Date'] = pd.to_datetime(df_clean['Date'])

    dominant_trees = []
    # Group by plot and date to find the local height dominants
    for _, group in df_clean.groupby(['Date', 'Spacing', 'Species', 'Plot']):
        threshold = group[height_col].quantile(0.80)
        top_20 = group[group[height_col] >= threshold]
        dominant_trees.append(top_20)

    # Return the unaggregated individual dominant trees
    return pd.concat(dominant_trees, ignore_index=True)


# ==============================================================================
# HELPER: Fit GAM to Individual Trees & Generate Smooth Plotting Curve
# ==============================================================================
def fit_true_gam(dates_series, values_series, n_splines=6):
    """
    Fits a GAM to raw individual data.
    Returns a smooth high-resolution X array, predicted Y, and +/- 1 SE band.
    """
    x_num = mdates.date2num(dates_series.values)
    y = values_series.values

    # GAM splines are limited by unique X values (dates), not total trees
    unique_dates = len(np.unique(x_num))

    if unique_dates > 3:
        X = x_num.reshape(-1, 1)
        actual_splines = min(n_splines, unique_dates)

        try:
            gam = LinearGAM(s(0, n_splines=actual_splines)).fit(X, y)

            # Generate 100 evenly spaced points for a perfectly smooth curve
            x_smooth = np.linspace(x_num.min(), x_num.max(), 100)
            X_smooth = x_smooth.reshape(-1, 1)

            y_pred = gam.predict(X_smooth)

            # +/- 1 Standard Error based on individual tree variance
            # A 68.27% confidence interval in a normal distribution is exactly +/- 1 SE
            y_intervals = gam.confidence_intervals(X_smooth, width=0.6827)

            # Convert smooth X back to dates for matplotlib
            return mdates.num2date(x_smooth), y_pred, y_intervals
        except Exception:
            pass # Fallback to raw if singular matrix

    return None, None, None


# ==============================================================================
# MASTER 2-METRIC PLOTTING FUNCTION
# ==============================================================================
def plot_dominant_2x4(df_source,
                      top_metric_col, top_metric_name, top_ylabel,
                      bot_metric_col, bot_metric_name, bot_ylabel,
                      csv_prefix, n_splines=6):

    # Filter culture == Single
    df_source = df_source[df_source["Culture"] == "Single"].copy()

    # 1. Extract Raw Dominant Trees ONCE based strictly on height
    df_dom = filter_dominant_by_height(df_source, 'Calibrated_Height')

    # Exporting the raw dominant dataset (useful for paper methodology)
    df_dom.to_csv(f"dominant_{csv_prefix}_raw.csv", index=False)

    spacings = sorted(df_dom['Spacing'].unique())

    # sharey='row' automatically handles hiding inner y-tick labels
    fig, axes = plt.subplots(nrows=2, ncols=len(spacings),
                             figsize=(7.5, 4),
                             sharex=True, sharey='row')

    row_meta = [
        (top_metric_col, top_metric_name, top_ylabel),
        (bot_metric_col, bot_metric_name, bot_ylabel),
    ]

    for row_idx, (metric_col, metric_name, ylabel) in enumerate(row_meta):
        for col_idx, spacing in enumerate(spacings):
            ax = axes[row_idx, col_idx]

            # Filter to spacing and drop NAs for the specific metric being plotted
            spacing_data = df_dom[df_dom['Spacing'] == spacing].dropna(subset=[metric_col])

            for species in legend_order_2:
                sp_data = spacing_data[spacing_data['Species'] == species]
                if sp_data.empty:
                    continue

                color = species_colors[species]
                z_order = 3 if species == 'Mixed' else 2

                # 2. Fit the True GAM
                x_dates, mean_gam, se_gam = fit_true_gam(sp_data['Date'], sp_data[metric_col], n_splines)

                if x_dates is not None:
                    # GAM Succeeded: Plot smooth line and +/- 1 SE band
                    lower_bound = se_gam[:, 0]
                    upper_bound = se_gam[:, 1]

                    ax.plot(x_dates, mean_gam, color=color, linewidth=1.5,
                            linestyle='-', label=species, zorder=z_order)

                    ax.fill_between(x_dates, lower_bound, upper_bound,
                                    color=color, alpha=0.15, zorder=z_order - 1)
                else:
                    # GAM Failed (Too few dates): Fallback to calculating mean/SE on the fly
                    fallback_agg = sp_data.groupby('Date')[metric_col].agg(['mean', 'std', 'count']).reset_index()
                    fallback_agg['se'] = fallback_agg['std'] / np.sqrt(fallback_agg['count'])

                    ax.errorbar(
                        fallback_agg['Date'], fallback_agg['mean'],
                        yerr=fallback_agg['se'],
                        color=color, linewidth=1.0, linestyle='-',
                        marker='o', markersize=4, capsize=2, capthick=0.8,
                        elinewidth=0.8, label=species, zorder=z_order
                    )

            # Spacing bubble — top row only
            if row_idx == 0:
                ax.legend([], [], title=f"{int(spacing)}x{int(spacing)}m",
                          loc='upper center', labelspacing=0, borderpad=0.3)

            # Clean Y-axis logic: only label the first column
            if col_idx == 0:
                ax.set_ylabel(ylabel, fontweight='bold')

            # X-axis logic
            ax.xaxis.set_major_locator(mdates.MonthLocator(interval=3))
            if row_idx == 1:
                ax.xaxis.set_major_formatter(mdates.DateFormatter('%b %Y'))
                ax.tick_params(axis='x', rotation=45, labelbottom=True)
                for tick in ax.get_xticklabels():
                    tick.set_horizontalalignment('right')
            else:
                ax.tick_params(axis='x', labelbottom=False)

        # Apply y-limit for the entire row AFTER all columns are plotted
        axes[row_idx, 0].set_ylim(bottom=0)

    # Master legend
    handles, labels = axes[0, 0].get_legend_handles_labels()
    by_label = dict(zip(labels, handles))
    fig.legend(by_label.values(), by_label.keys(),
               loc='lower center', bbox_to_anchor=(0.5, 0.91),
               ncol=len(by_label))

    fig.suptitle(
        f"Dominant {top_metric_name} and {bot_metric_name} by spacing and species",
        fontweight='bold', y=1.04
    )

    add_panel_labels([axes[0,0],axes[0,1],axes[0,2],axes[0,3],axes[1,0],axes[1,1],axes[1,2],axes[1,3]], 8)

    plt.tight_layout(rect=[0, 0, 1.0, 0.98])
    plt.subplots_adjust(hspace=0.1, wspace=0.22)
    plt.savefig(f'/content/drive/My Drive/EucVision/Figures/{csv_prefix}_trueGAM_2x4.png', dpi=300, bbox_inches='tight')
    plt.show()

# ==============================================================================
# EXECUTE
# ==============================================================================

# Ensure your global variables (like legend_order_2, species_colors, tab10_colors)
# and your dataframe 'df' are loaded in memory before running this!

# Figure 1: Crown Area (top) + Tree Height (bottom)
plot_dominant_2x4(
    df_source=df,
    top_metric_col='Crown_Area',       top_metric_name='crown area',  top_ylabel='Mean crown area\n per tree (m²)',
    bot_metric_col='Calibrated_Height', bot_metric_name='calibrated height', bot_ylabel='Mean calibrated height\n per tree (m)',
    csv_prefix='crown_height',
    n_splines=12
)

# Figure 2: Crown Volume (top) + Stem Volume (bottom)
# (Make sure 'Crown_Volume' and 'Stem_Volume' match your exact column names in df)
plot_dominant_2x4(
    df_source=df,
    top_metric_col='Crown_Volume',  top_metric_name='crown volume',  top_ylabel='Mean cone-equivalent\ncrown volume per tree (m³)',
    bot_metric_col='Stem_Volume', bot_metric_name='stem volume', bot_ylabel='Mean cone-equivalent\nstem volume per tree (m³)',
    csv_prefix='Stem_Crown_Volume',
    n_splines=12
)
```

```python
# ==============================================================================
# @title Slenderness & Gini Index
# 3 rows × 4 spacing columns
# ==============================================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.dates as mdates
import matplotlib.cm as cm
from scipy import stats
from pygam import LinearGAM, s

# ==============================================================================
# 1. LOAD & CLEAN DATA
# ==============================================================================
start_date = pd.to_datetime('2025-09-01')

file_path_df_master = '/content/drive/My Drive/EucVision/UAV_Master_Dataset_26-06-2026.csv'
df_master = pd.read_csv(file_path_df_master, low_memory=False)
df_master['Date'] = pd.to_datetime(df_master['Date'], errors='coerce')

borrowed_dates = pd.to_datetime(['2025-11-14', '2026-03-16', '2026-04-08',
                                 '2026-04-13', '2026-04-29'])
df_master.loc[df_master['Date'].isin(borrowed_dates), 'Crown_Area_m2'] = np.nan

df_master = df_master[df_master['Date'] >= start_date]
df_master = df_master[df_master['Death_Date'] == 'Alive'].copy()

# ==============================================================================
# 2. DERIVE SLENDERNESS INDEX
# ==============================================================================
df_master['Crown_Radius_m'] = np.sqrt(df_master['Crown_Area_m2'] / np.pi)
df_master['Slenderness_Index'] = (
    df_master['Calibrated_Height_m'] / df_master['Crown_Radius_m']
)
df_master.loc[df_master['Crown_Radius_m'] < 0.01, 'Slenderness_Index'] = np.nan

# ==============================================================================
# 3. HELPERS
# ==============================================================================
def filter_dominant_by_height_mixed_aware(df_source, height_col='Calibrated_Height_m'):
    """
    Top 20% of trees per date/spacing/species group based STRICTLY ON HEIGHT.
    """
    df_clean = df_source.dropna(
        subset=[height_col, 'Species', 'Spacing', 'Date', 'Culture']
    ).copy()
    dominant_frames = []

    # Pure stands — Dominance evaluated within the species monoculture
    df_pure = df_clean[df_clean['Culture'] == 'Single']
    for (date, spacing, species), group in df_pure.groupby(['Date', 'Spacing', 'Species']):
        threshold = group[height_col].quantile(0.80)
        dominant_frames.append(group[group[height_col] >= threshold])

    # Mixed stands — Dominance evaluated across the whole mixed plot
    df_mix = df_clean[df_clean['Culture'] == 'Mix']
    for (date, spacing), group in df_mix.groupby(['Date', 'Spacing']):
        threshold = group[height_col].quantile(0.80)
        top_20 = group[group[height_col] >= threshold].copy()
        top_20['Species'] = 'Mixed' # Relabel for the plotting legend
        dominant_frames.append(top_20)

    return pd.concat(dominant_frames, ignore_index=True)

def fit_true_gam(dates_series, values_series, n_splines):
    """Fits a GAM and returns smooth X, Y, and ±1 SE bands."""
    x_num = mdates.date2num(pd.to_datetime(dates_series).values)
    y = values_series.values
    unique_dates = len(np.unique(x_num))

    if unique_dates > 3:
        X = x_num.reshape(-1, 1)
        actual_splines = min(n_splines, unique_dates)
        try:
            gam = LinearGAM(s(0, n_splines=actual_splines)).fit(X, y)
            x_smooth = np.linspace(x_num.min(), x_num.max(), 100)
            y_pred = gam.predict(x_smooth.reshape(-1, 1))
            y_intervals = gam.confidence_intervals(x_smooth.reshape(-1, 1), width=0.6827)
            return mdates.num2date(x_smooth), y_pred, y_intervals
        except Exception:
            pass
    return None, None, None


def gini_coefficient(values):
    """
    Compute the Gini coefficient for an array of non-negative values.
    Returns NaN if fewer than 2 valid values.
    """
    arr = np.array(values, dtype=float)
    arr = arr[~np.isnan(arr) & (arr >= 0)]
    n = len(arr)
    if n < 2:
        return np.nan
    arr = np.sort(arr)
    # Standard formula: G = (2 * sum(i * x_i)) / (n * sum(x_i)) - (n+1)/n
    idx = np.arange(1, n + 1)
    return (2 * np.sum(idx * arr) / (n * np.sum(arr))) - (n + 1) / n


def compute_gini_timeseries(df_source, metric_col):
    """
    Compute per-date Gini coefficient across the WHOLE population
    for each (Spacing, Species/Mixed) group.
    Returns a dict keyed by spacing, each value a dict keyed by species
    containing a DataFrame with columns [Date, Gini].
    """
    df_clean = df_source.dropna(subset=[metric_col, 'Spacing', 'Date', 'Culture']).copy()
    result = {}

    spacings = sorted(df_clean['Spacing'].unique())
    for spacing in spacings:
        result[spacing] = {}
        sp_data = df_clean[df_clean['Spacing'] == spacing]

        # Pure stands — per species
        pure = sp_data[sp_data['Culture'] == 'Single']
        for species, grp in pure.groupby('Species'):
            ts = (grp.groupby('Date')[metric_col]
                     .apply(gini_coefficient)
                     .reset_index()
                     .rename(columns={metric_col: 'Gini'}))
            result[spacing][species] = ts

        # Mixed stands — all trees in mixed plots together
        mixed = sp_data[sp_data['Culture'] == 'Mix']
        if not mixed.empty:
            ts = (mixed.groupby('Date')[metric_col]
                       .apply(gini_coefficient)
                       .reset_index()
                       .rename(columns={metric_col: 'Gini'}))
            result[spacing]['Mixed'] = ts

    return result

# ==============================================================================
# 4. PREPARE DATA FOR EACH ROW
# ==============================================================================

# Row 0 — Slenderness: dominant trees only (Determined strictly by HEIGHT)
df_dominant_height = filter_dominant_by_height_mixed_aware(df_master, 'Calibrated_Height_m')
# Now that we have the dominant trees, isolate their slenderness values
si_raw_dominant = df_dominant_height.dropna(subset=['Slenderness_Index']).copy()

# Row 1 — Gini: whole population, on Crown_Area_m2
gini_data = compute_gini_timeseries(df_master, 'Crown_Area_m2')

# Row 2 — Allometry: whole population (scatter handled directly in plot loop)

# ==============================================================================
# 5. MAIN PLOT  —  3 rows × 4 spacing columns
# ==============================================================================
spacings = sorted(df_master['Spacing'].unique())
n_cols   = len(spacings)

fig, axes = plt.subplots(nrows=3, ncols=n_cols,
                         figsize=(MW, 6),   # taller to accommodate 3 rows
                         sharey='row',
                         sharex='row')

# ── ROW 0: Slenderness Index (GAM on dominant trees) ─────────────────────────
for col_idx, spacing in enumerate(spacings):
    ax = axes[0, col_idx]
    sp_data = si_raw_dominant[si_raw_dominant['Spacing'] == spacing].sort_values('Date')

    for species in legend_order:
        s_data = sp_data[sp_data['Species'] == species].dropna(subset=['Slenderness_Index'])
        if s_data.empty:
            continue

        color   = species_colors.get(species, 'black')
        z_order = 3 if species == 'Mixed' else 2

        x_dates, mean_gam, se_gam = fit_true_gam(
            s_data['Date'], s_data['Slenderness_Index'], n_splines=6
        )

        if x_dates is not None:
            ax.plot(x_dates, mean_gam, color=color, linewidth=1.5,
                    linestyle='-', label=species, zorder=z_order)
            ax.fill_between(x_dates, se_gam[:, 0], se_gam[:, 1],
                            color=color, alpha=0.15, zorder=z_order - 1)
        else:
            fallback = (s_data.groupby('Date')['Slenderness_Index']
                              .agg(['mean', 'std', 'count']).reset_index())
            fallback['se'] = fallback['std'] / np.sqrt(fallback['count'])
            ax.errorbar(fallback['Date'], fallback['mean'], yerr=fallback['se'],
                        color=color, linewidth=1.0, linestyle='-',
                        marker='o', markersize=4)

    ax.legend([], [], title=f"{int(spacing)}x{int(spacing)}m",
              loc='upper center', labelspacing=0, borderpad=0.3)

    if col_idx == 0:
        ax.set_ylabel('Slenderness Index\n(Height / Crown Radius)', fontweight='bold')
    else:
        ax.set_ylabel('')
        ax.tick_params(axis='y', length=0)

    ax.margins(y=0.20)
    ax.xaxis.set_major_locator(mdates.MonthLocator(interval=3))
    ax.xaxis.set_major_formatter(mdates.DateFormatter('%b %Y'))
    ax.tick_params(axis='x', rotation=45, labelbottom=True)
    for tick in ax.get_xticklabels():
        tick.set_horizontalalignment('right')

# ── ROW 1: Gini Coefficient (whole population, Crown Area) ───────────────────
for col_idx, spacing in enumerate(spacings):
    ax = axes[1, col_idx]
    spacing_gini = gini_data.get(spacing, {})

    for species in legend_order:
        ts = spacing_gini.get(species)
        if ts is None or ts.empty:
            continue

        ts = ts.dropna(subset=['Gini']).sort_values('Date')
        if ts.empty:
            continue

        color   = species_colors.get(species, 'black')
        z_order = 3 if species == 'Mixed' else 2

        # Fit GAM to Gini time series if enough points
        x_dates, mean_gam, se_gam = fit_true_gam(
            ts['Date'], ts['Gini'], n_splines=6
        )

        if x_dates is not None:
            ax.plot(x_dates, mean_gam, color=color, linewidth=1.5,
                    linestyle='-', label=species, zorder=z_order)
            ax.fill_between(x_dates, se_gam[:, 0], se_gam[:, 1],
                            color=color, alpha=0.15, zorder=z_order - 1)
        else:
            # Fallback: raw Gini points connected by line
            ax.plot(ts['Date'], ts['Gini'], color=color, linewidth=1.0,
                    linestyle='-', marker='o', markersize=3, label=species,
                    zorder=z_order)

    # Reference lines: Gini interpretation thresholds
    ax.axhline(0.0, color='grey', linewidth=0.5, linestyle='--', alpha=0.5)
    ax.axhline(0.5, color='grey', linewidth=0.5, linestyle=':', alpha=0.5)

    # ax.legend([], [], title=f"{int(spacing)} m",
    #          loc='upper left', labelspacing=0, borderpad=0.3)

    if col_idx == 0:
        ax.set_ylabel('Gini Coefficient\n(Crown Area Inequality)', fontweight='bold')
    else:
        ax.set_ylabel('')
        ax.tick_params(axis='y', length=0)

    ax.set_ylim(0, 0.75)   # Gini is bounded [0,1]; 0.75 gives headroom
    ax.margins(y=0.10)
    ax.xaxis.set_major_locator(mdates.MonthLocator(interval=3))
    ax.xaxis.set_major_formatter(mdates.DateFormatter('%b %Y'))
    ax.tick_params(axis='x', rotation=45, labelbottom=True)
    for tick in ax.get_xticklabels():
        tick.set_horizontalalignment('right')

# ── ROW 2: Allometric Scatter (Crown Area vs Height, whole population) ────────
for col_idx, spacing in enumerate(spacings):
    ax = axes[2, col_idx]

    mask    = (df_master['Spacing'].astype(float) == spacing)
    plot_df = df_master[mask].dropna(
        subset=['Crown_Area_m2', 'Calibrated_Height_m', 'Species', 'Date']
    ).copy()

    for species in legend_order:
        if species == 'Mixed':
            sp_df = plot_df[plot_df['Culture'] == 'Mix'].copy()
        else:
            sp_df = plot_df[
                (plot_df['Species'] == species) & (plot_df['Culture'] == 'Single')
            ].copy()

        if sp_df.empty:
            continue

        color  = species_colors.get(species, 'black')
        x_vals = sp_df['Crown_Area_m2'].values
        y_vals = sp_df['Calibrated_Height_m'].values

        valid  = (x_vals > 0) & (y_vals > 0)
        x_v, y_v = x_vals[valid], y_vals[valid]

        if len(x_v) > 5:
            slope, intercept, r, p, _ = stats.linregress(np.log(x_v), np.log(y_v))
            A = np.exp(intercept)
            x_line = np.linspace(x_v.min(), x_v.max(), 100)
            ax.plot(x_line, A * (x_line ** slope), color=color, linewidth=1.5,
                    linestyle='-', label=species, zorder=3)

    if col_idx == 0:
        ax.set_ylabel('Height (m)', fontweight='bold')
    else:
        ax.set_ylabel('')
        ax.tick_params(axis='y', length=0)

    ax.set_xlabel('Crown Area (m²)', fontweight='bold')
    ax.margins(y=0.10)

# ==============================================================================
# 6. MASTER LEGEND & TITLES
# ==============================================================================
by_label = {}
for row in range(3):
    h, l = axes[row, 0].get_legend_handles_labels()
    by_label.update(dict(zip(l, h)))

ordered_handles = [by_label[sp] for sp in legend_order if sp in by_label]
ordered_labels  = [sp           for sp in legend_order if sp in by_label]

fig.legend(ordered_handles, ordered_labels,
           loc='lower center', bbox_to_anchor=(0.5, 0.94),
           ncol=len(ordered_labels))

fig.suptitle(
    'Slenderness Index, Crown Area Inequality (Gini) & Allometry by Spacing',
    fontweight='bold', y=1.02
)

add_panel_labels(axes.flatten(), 12)

plt.tight_layout(rect=[0.01, 0, 1.0, 0.97])
plt.subplots_adjust(hspace=0.40, wspace=0.22)
plt.savefig('/content/drive/My Drive/EucVision/Figures/Slenderness_And_Gini.png')
plt.show()
```

```python
# @title Growth Rate

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.dates as mdates
from pygam import LinearGAM, s

# ==============================================================================
# 1. LOAD & CLEAN DATA
# ==============================================================================
# Define strict start date (Bypasses the winter data gap)
start_date = pd.to_datetime('2025-10-31')

# --- Drone Data ---
file_path_df_master = '/content/drive/My Drive/EucVision/UAV_Master_Dataset_26-06-2026.csv'
df_master = pd.read_csv(file_path_df_master, low_memory=False)

df_master['Date'] = pd.to_datetime(df_master['Date'], errors='coerce')

# --- Mask borrowed 2D Crown Area Data ---
borrowed_dates = pd.to_datetime(['2025-11-14', '2026-03-16', '2026-04-08', '2026-04-13', '2026-04-29'])
df_master.loc[df_master['Date'].isin(borrowed_dates), 'Crown_Area_m2'] = np.nan

df_master = df_master[df_master['Date'] >= start_date].copy()

if 'Death_Date' in df_master.columns:
    df_master['Parsed_Death_Date'] = pd.to_datetime(df_master['Death_Date'], format='%d-%m-%Y', errors='coerce')
    df_master['Is_Dead'] = df_master['Parsed_Death_Date'].notna() & (df_master['Date'] >= df_master['Parsed_Death_Date'])
    df_master.loc[df_master['Is_Dead'], ['Crown_Area_m2', 'Calibrated_Height_m']] = np.nan
    df_master = df_master.drop(columns=['Parsed_Death_Date', 'Is_Dead'])

# ==============================================================================
# 2. HELPER: Extract Dominant Raw Individuals (Strictly by Height)
# ==============================================================================
def filter_dominant_by_height(df_source, height_col='Calibrated_Height_m'):
    """
    For each Date/Spacing/Species/Plot, isolates the top-20% dominant trees
    based strictly on HEIGHT.
    """
    df_clean = df_source[df_source['Culture'] == 'Single'].dropna(
    subset=[height_col, 'Species', 'Spacing', 'Plot', 'Date']
    ).copy()

    dominant_trees = []
    # Group by Date, Spacing, and Species to find local height dominants
    for _, group in df_clean.groupby(['Date', 'Spacing', 'Species', 'Plot']):
        threshold = group[height_col].quantile(0.80)
        dominant_trees.append(group[group[height_col] >= threshold])

    return pd.concat(dominant_trees, ignore_index=True)

# ==============================================================================
# 3. HELPER: GAM Derivative
# ==============================================================================
def safe_gam_derivative(dates_series, values_series, n_splines):
    """
    Fits a GAM to raw individual data and calculates the WEEKLY growth rate.
    Strictly uses the user-defined n_splines with no automatic capping.
    """
    x_num = mdates.date2num(pd.to_datetime(dates_series).values)
    y = values_series.values

    # Basic check just to prevent total matrix collapse on empty plots
    if len(np.unique(x_num)) > 3:
        X = x_num.reshape(-1, 1)

        try:
            # 100% manual spline control
            gam = LinearGAM(s(0, n_splines=n_splines)).fit(X, y)

            x_smooth = np.linspace(x_num.min(), x_num.max(), 100)
            y_pred = gam.predict(x_smooth.reshape(-1, 1))

            daily_derivative = np.gradient(y_pred, x_smooth)
            weekly_derivative = daily_derivative * 7

            return mdates.num2date(x_smooth), weekly_derivative
        except Exception:
            pass

    return mdates.num2date(x_num), np.zeros(len(y))

# ==============================================================================
# 4. MASTER 2x4 PLOTTING FUNCTION
# ==============================================================================
def plot_growth_2x4(df_source,
                    top_metric_col, top_metric_name, top_ylabel,
                    bot_metric_col, bot_metric_name, bot_ylabel,
                    csv_prefix,
                    user_spline_count):

    # 1. Extract dominant trees ONCE based strictly on height
    df_dom = filter_dominant_by_height(df_source, 'Calibrated_Height_m')

    # Exporting the raw dominant dataset
    df_dom.to_csv(f"dominant_{csv_prefix}_raw.csv", index=False)

    spacings = sorted(df_dom['Spacing'].unique())

    fig, axes = plt.subplots(nrows=2, ncols=len(spacings),
                             figsize=(7.5, 4),
                             sharex=True, sharey='row')

    row_meta = [
        (top_metric_name, top_ylabel, top_metric_col),
        (bot_metric_name, bot_ylabel, bot_metric_col),
    ]

    for row_idx, (metric_name, ylabel, metric_col) in enumerate(row_meta):
        for col_idx, spacing in enumerate(spacings):
            ax = axes[row_idx, col_idx]

            # Filter to spacing and drop NAs for the specific metric being plotted
            spacing_data = df_dom[df_dom['Spacing'] == spacing].dropna(subset=[metric_col])

            ax.axhline(0, color='#555555', linewidth=1.2, linestyle='--', zorder=1, alpha=0.9)

            for species in legend_order:
                sp_data = spacing_data[spacing_data['Species'] == species]
                if sp_data.empty: continue

                color = species_colors.get(species, 'black')
                z_order = 3 if species == 'Mixed' else 2

                # Passing your manually tuned spline count directly to the fitter
                x_out, y_out = safe_gam_derivative(sp_data['Date'], sp_data[metric_col], n_splines=user_spline_count)

                ax.plot(x_out, y_out, color=color, linewidth=1.5, linestyle='-',
                        label=species, zorder=z_order)

            if row_idx == 0:
                ax.legend([], [], title=f"{int(spacing)}x{int(spacing)}m",
                          loc='upper center', labelspacing=0, borderpad=0.3)

            if col_idx == 0:
                ax.set_ylabel(ylabel, fontweight='bold')
            else:
                ax.set_ylabel('')
                ax.tick_params(axis='y', length=0)

            ax.set_xlabel('')
            ax.set_title('')
            ax.margins(y=0.20)

            if row_idx == 1:
                ax.xaxis.set_major_locator(mdates.MonthLocator(interval=3))
                ax.xaxis.set_major_formatter(mdates.DateFormatter('%b %Y'))
                ax.tick_params(axis='x', rotation=45, labelbottom=True)
                for tick in ax.get_xticklabels():
                    tick.set_horizontalalignment('right')
            else:
                ax.tick_params(axis='x', labelbottom=False)

    handles, labels = axes[0, 0].get_legend_handles_labels()
    by_label = dict(zip(labels, handles))

    ordered_handles = [by_label[sp] for sp in legend_order if sp in by_label]
    ordered_labels = [sp for sp in legend_order if sp in by_label]

    fig.legend(ordered_handles, ordered_labels,
               loc='lower center', bbox_to_anchor=(0.5, 0.91),
               ncol=len(ordered_labels))

    fig.suptitle(f"Dominant crown area and calibrated height growth rate by spacing and species",
                 fontweight='bold', y=1.03)
    add_panel_labels(axes.flatten(), 8)

    plt.tight_layout(rect=[0, 0, 1.0, 1.0])
    plt.subplots_adjust(hspace=0.1, wspace=0.22)
    plt.savefig(f'/content/drive/My Drive/EucVision/Figures/{csv_prefix}_growth_2x4_GAM.png', dpi=300, bbox_inches='tight')
    plt.show()

# ==============================================================================
# 5. EXECUTE THE 2x4 GRAPH
# ==============================================================================

# ⬇️⬇️⬇️ TUNE YOUR SPLINES HERE ⬇️⬇️⬇️
MANUAL_SPLINE_COUNT = 7

plot_growth_2x4(
    df_source=df_master,
    top_metric_col='Crown_Area_m2',
    top_metric_name='Crown Area',
    top_ylabel='Mean area\n growth rate\n per tree (m²/week)',
    bot_metric_col='Calibrated_Height_m',
    bot_metric_name='Tree Height',
    bot_ylabel='Mean calibrated height\n growth rate\n per tree (m/week)',
    csv_prefix='velocity_crown_height',
    user_spline_count=MANUAL_SPLINE_COUNT  # Feeds straight into the plotter
)
```

```python
# @title PCA
# (Time-Series 3x2 Grid: Crown, Height, & Combined 2D Mass)
# Sized for MW (19.0 cm / 7.48 in) — MSc Word Thesis

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import matplotlib as mpl
import seaborn as sns
import matplotlib.transforms as transforms
from matplotlib.patches import Ellipse
from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler
import warnings
warnings.filterwarnings('ignore')

# Assuming 'df', 'species_colors', 'legend_order', and 'MW' are already loaded/defined

# ---------------------------------------------------------
# 0. DERIVE SCALED SIZES FROM RCPARAMS (no hardcoding)
# ---------------------------------------------------------
_base   = mpl.rcParams['font.size']         # e.g. 9 pt
_title  = mpl.rcParams['axes.titlesize']    # e.g. 10 pt
_label  = mpl.rcParams['axes.labelsize']    # e.g. 9 pt
_tick   = mpl.rcParams['xtick.labelsize']   # e.g. 8 pt
_leg    = mpl.rcParams['legend.fontsize']   # e.g. 8 pt
_legtit = mpl.rcParams['legend.title_fontsize']  # e.g. 10 pt

# Scale factor relative to the original 16 cm / FW figure
_sf = MW / 6.30  # ≈ 1.19 — used to scale non-font visual elements

# ---------------------------------------------------------
# 1. CONFIDENCE ELLIPSE HELPER
# ---------------------------------------------------------
def confidence_ellipse(x, y, ax, n_std=2.0, facecolor='none', **kwargs):
    if x.size != y.size:
        raise ValueError("x and y must be the same size")
    cov = np.cov(x, y)
    pearson = cov[0, 1] / np.sqrt(cov[0, 0] * cov[1, 1])
    ell_radius_x = np.sqrt(1 + pearson)
    ell_radius_y = np.sqrt(1 - pearson)
    ellipse = Ellipse((0, 0),
                      width=ell_radius_x * 2,
                      height=ell_radius_y * 2,
                      facecolor=facecolor, **kwargs)
    scale_x = np.sqrt(cov[0, 0]) * n_std
    mean_x  = np.mean(x)
    scale_y = np.sqrt(cov[1, 1]) * n_std
    mean_y  = np.mean(y)
    transf = (transforms.Affine2D()
              .rotate_deg(45)
              .scale(scale_x, scale_y)
              .translate(mean_x, mean_y))
    ellipse.set_transform(transf + ax.transData)
    return ax.add_patch(ellipse)

# ---------------------------------------------------------
# 2. RUN ALL 3 PCA CALCULATIONS UPFRONT
# ---------------------------------------------------------

# Toggle to test sensitivity to the fillna(0) artifact:
# True  = only trees measured on every survey date (no zero-filling)
# False = original behaviour (missing dates -> 0)
REQUIRE_COMPLETE_COVERAGE = True

def build_pivot(value_col):
    pivot = df.pivot_table(index='Tree_ID', columns='Date', values=value_col)
    n_total = len(pivot)
    if REQUIRE_COMPLETE_COVERAGE:
        pivot = pivot.dropna(axis=0, how='any')
    else:
        pivot = pivot.fillna(0)
    print(f"{value_col}: {len(pivot)} / {n_total} trees retained "
          f"({'complete coverage only' if REQUIRE_COMPLETE_COVERAGE else 'gaps filled with 0'})")
    return pivot

# --- PCA 1: Crown Area ---
pivot_crown = build_pivot('Crown_Area')
X_crown = StandardScaler().fit_transform(pivot_crown.values)
pca_crown = PCA(n_components=2)
pca_df_crown = pd.DataFrame(
    data=pca_crown.fit_transform(X_crown),
    columns=['PC1', 'PC2'],
    index=pivot_crown.index
)

# --- PCA 2: Adjusted Height ---
pivot_height = build_pivot('Calibrated_Height')
X_height = StandardScaler().fit_transform(pivot_height.values)
pca_height = PCA(n_components=2)
pca_df_height = pd.DataFrame(
    data=pca_height.fit_transform(X_height),
    columns=['PC1', 'PC2'],
    index=pivot_height.index
)

# --- PCA 3: Combined (Crown Area + Height ONLY) ---
# Restrict to trees with complete coverage in BOTH metrics, so the combined
# analysis doesn't reintroduce zeros through the back door via concat.
if REQUIRE_COMPLETE_COVERAGE:
    common_ids = pivot_crown.index.intersection(pivot_height.index)
else:
    common_ids = pivot_crown.index.union(pivot_height.index)

pivot_crown_tagged  = pivot_crown.reindex(common_ids).copy()
pivot_height_tagged = pivot_height.reindex(common_ids).copy()
pivot_crown_tagged.columns  = ['Crown_'  + str(c) for c in pivot_crown_tagged.columns]
pivot_height_tagged.columns = ['Height_' + str(c) for c in pivot_height_tagged.columns]

pivot_combined = pd.concat([pivot_crown_tagged, pivot_height_tagged], axis=1)
if not REQUIRE_COMPLETE_COVERAGE:
    pivot_combined = pivot_combined.fillna(0)
print(f"Combined: {len(pivot_combined)} trees retained")

X_combined = StandardScaler().fit_transform(pivot_combined.values)
pca_combined = PCA(n_components=2)
pca_df_combined = pd.DataFrame(
    data=pca_combined.fit_transform(X_combined),
    columns=['PC1', 'PC2'],
    index=pivot_combined.index
)

# Smart axis flip: ensure Height always points UP (positive PC2)
coeff = np.transpose(pca_combined.components_)
height_idx = [i for i, c in enumerate(pivot_combined.columns) if 'Height' in c]
height_vec_pre = np.sum(coeff[height_idx, :], axis=0)
if height_vec_pre[1] < 0:
    pca_df_combined['PC2']               = pca_df_combined['PC2'] * -1
    pca_combined.components_[1, :]       = pca_combined.components_[1, :] * -1

# --- Join Metadata ---
metadata        = df[['Tree_ID', 'Species', 'Spacing']].drop_duplicates().set_index('Tree_ID')
pca_df_crown    = pca_df_crown.join(metadata)
pca_df_height   = pca_df_height.join(metadata)
pca_df_combined = pca_df_combined.join(metadata)

# ---------------------------------------------------------
# 3. SETUP 3×2 GRID  —  MW wide, proportional height
# ---------------------------------------------------------
fig, axes = plt.subplots(nrows=3, ncols=2, figsize=(FW, 9))
fig.suptitle(
    'Growth dynamics PCA with 95% confidence ellipses',
    fontweight='bold', y=1
)
axes = axes.flatten()

plot_configs = [
    {'df': pca_df_crown,    'pca_model': pca_crown,    'col': 'Species', 'title': 'Crown area by species'},
    {'df': pca_df_crown,    'pca_model': pca_crown,    'col': 'Spacing', 'title': 'Crown area by spacing'},
    {'df': pca_df_height,   'pca_model': pca_height,   'col': 'Species', 'title': 'Calibrated height by species'},
    {'df': pca_df_height,   'pca_model': pca_height,   'col': 'Spacing', 'title': 'Calibrated height by spacing'},
    {'df': pca_df_combined, 'pca_model': pca_combined, 'col': 'Species', 'title': 'Crown area and calibrated height by species'},
    {'df': pca_df_combined, 'pca_model': pca_combined, 'col': 'Spacing', 'title': 'Crown area and calibrated height by spacing'},
]

for i, config in enumerate(plot_configs):
    ax       = axes[i]
    plot_df  = config['df'].dropna(subset=[config['col']]).copy()
    pca_model = config['pca_model']
    col       = config['col']

    if col == 'Spacing':
        plot_df[col] = plot_df[col].apply(lambda x: f"{int(x)}x{int(x)}m")

    unique_vals = sorted(plot_df[col].unique())

    if col == 'Species':
        palette   = species_colors
        hue_order = legend_order
    else:  # Spacing
        pal_colors = sns.color_palette("viridis", n_colors=len(unique_vals))
        palette    = dict(zip(unique_vals, pal_colors))
        hue_order  = unique_vals

    show_legend = i < 2   # only panels (a) and (b)

    # A. Scatter points — dot size scales with figure area
    sns.scatterplot(
        data=plot_df, x='PC1', y='PC2', hue=col,
        alpha=0.4, s=int(10 * _sf),
        palette=palette, hue_order=hue_order, ax=ax,
        legend=show_legend
    )

    # B. Confidence ellipses — linewidth scales with figure width
    for val in unique_vals:
        subset = plot_df[plot_df[col] == val]
        if len(subset) > 5:
            confidence_ellipse(
                subset['PC1'], subset['PC2'], ax, n_std=2,
                edgecolor=palette[val], linewidth=1.5 * _sf
            )

    # C. Trajectory arrows (combined PCA panels only)
    if pca_model is pca_combined:
        coeff      = np.transpose(pca_model.components_)
        crown_idx  = [j for j, c in enumerate(pivot_combined.columns) if 'Crown'  in c]
        height_idx = [j for j, c in enumerate(pivot_combined.columns) if 'Height' in c]

        crown_vec  = np.sum(coeff[crown_idx,  :], axis=0)
        height_vec = np.sum(coeff[height_idx, :], axis=0)

        max_scatter  = max(plot_df['PC1'].abs().max(), plot_df['PC2'].abs().max())
        max_vec      = max(np.linalg.norm(crown_vec), np.linalg.norm(height_vec))
        scale_factor = (max_scatter * 0.40) / max_vec

        def draw_trajectory(vec, label):
            vx, vy = vec[0] * scale_factor, vec[1] * scale_factor
            ax.arrow(
                0, 0, vx, vy,
                color='black', alpha=0.9,
                width=0.02 * _sf,
                head_width=0.15 * _sf,
                zorder=5
            )
            ax.text(
                vx * 1.25, vy * 1.25, label,
                color='black', fontweight='bold',
                ha='center', va='center',
                bbox=dict(facecolor='white', alpha=0.7, edgecolor='black', pad=3 * _sf),
                zorder=6
            )

        draw_trajectory(crown_vec,  'CROWN')
        draw_trajectory(height_vec, 'HEIGHT')

    # D. Formatting — all sizes pulled from rcParams
    ax.set_title(config['title'])
    ax.grid(True, alpha=0.3)
    ax.axhline(0, color='grey', linestyle='--', linewidth=1, alpha=0.5)
    ax.axvline(0, color='grey', linestyle='--', linewidth=1, alpha=0.5)

    pc1_var = pca_model.explained_variance_ratio_[0] * 100
    pc2_var = pca_model.explained_variance_ratio_[1] * 100
    ax.set_xlabel(f'PC1: Overall size ({pc1_var:.1f}%)')
    ax.set_ylabel(f'PC2: Growth timing ({pc2_var:.1f}%)')

    if show_legend:
        ax.legend(
            title=col,
            loc='upper right',
            frameon=True, fancybox=True, shadow=True,
            framealpha=0.95, facecolor='white'
        )

add_panel_labels(axes, 6)

plt.tight_layout(rect=[0, 0, 1, 1])
plt.savefig('/content/drive/My Drive/EucVision/Figures/Full_PCA.png')
plt.show()
```

```python
# @title Mixed-Effects Models

# =============================================================================
# EucVision — Six Mixed-Effects Models (LMM)
# Eucalyptus species × spacing trial, IMPACT OAL, Stellenbosch
#
# Three responses × two populations:
#   Population 1: All alive trees
#   Population 2: Dominant trees (top 20% height per Species × Spacing × Date)
#
#   Response 1: Calibrated_Height_m   — Gaussian, raw scale
#   Response 2: Crown_Area_m2         — Gaussian, log-transformed
#   Response 3: CA:H ratio            — Gaussian, log-transformed
#
# Reference levels:
#   Species  → Grandis (commercial benchmark)
#   Spacing  → 1×1 m  (densest treatment)
#
# SPECIES FILTER:
#   Edit EXCLUDE_SPECIES below to remove confounded species.
#   Cladocalyx excluded by default — heavily grazed by grey buck in 2025.
# =============================================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.lines import Line2D
import statsmodels.formula.api as smf
import statsmodels.api as sm
import scipy.stats as stats
import warnings
warnings.filterwarnings("ignore")

# ── 1. FILTER SETTINGS — edit here ───────────────────────────────────────────
START_DATE      = "2025-09-01"          # analysis start date
EXCLUDE_SPECIES = []        # species to exclude (grazing confound)
                                        # set to [] to include all species
REFERENCE_SPECIES = "Grandis"           # LMM reference level for species
REFERENCE_SPACING = 1                   # LMM reference level for spacing

# ── 2. Load & prepare data ────────────────────────────────────────────────────
df_raw = pd.read_csv(
    '/content/drive/My Drive/EucVision/UAV_Master_Dataset_26-06-2026.csv'
)
df_raw["Date"] = pd.to_datetime(df_raw["Date"])

t0 = pd.Timestamp(START_DATE)
df_raw["days"]   = (df_raw["Date"] - t0).dt.days
df_raw["Months"] = df_raw["days"] / 30.44
df_raw["Plot_ID"] = df_raw["Compartment"].astype(str) + "_" + df_raw["Plot"].astype(str)

# Filter: alive trees from start date
df_lmm = df_raw[
    (df_raw["Death_Date"] == "Alive") &
    (df_raw["Date"] >= t0)
].copy()

# Exclude specified species
if EXCLUDE_SPECIES:
    df_lmm = df_lmm[~df_lmm["Species"].isin(EXCLUDE_SPECIES)].copy()
    print(f"Excluded species: {EXCLUDE_SPECIES}")

# Exclude mixed cultures
df_lmm = df_lmm[df_lmm["Culture"] == "Single"].copy()

# Derived response: CA:H ratio
df_lmm["CAH"] = df_lmm["Crown_Area_m2"] / df_lmm["Calibrated_Height_m"]
df_lmm["log_Crown"] = np.log(df_lmm["Crown_Area_m2"].clip(lower=1e-6))
df_lmm["log_CAH"]   = np.log(df_lmm["CAH"].clip(lower=1e-6))

# Set reference levels using Categorical
df_lmm["Species"] = pd.Categorical(
    df_lmm["Species"],
    categories=[REFERENCE_SPECIES] + sorted(
        [s for s in df_lmm["Species"].unique() if s != REFERENCE_SPECIES]
    )
)
df_lmm["Spacing"] = pd.Categorical(
    df_lmm["Spacing"],
    categories=[REFERENCE_SPACING] + sorted(
        [s for s in df_lmm["Spacing"].unique() if s != REFERENCE_SPACING]
    )
)

print(f"\nData loaded: {len(df_lmm):,} observations, {df_lmm['Tree_ID'].nunique():,} trees")
print(f"Species: {list(df_lmm['Species'].cat.categories)}")
print(f"Spacings: {list(df_lmm['Spacing'].cat.categories)}")
print(f"Months range: {df_lmm['Months'].min():.1f} to {df_lmm['Months'].max():.1f}")

# ── 3. Dominant tree filter (top 20% height per Species × Spacing × Date) ────
df_lmm["dom_threshold"] = df_lmm.groupby(
    ["Species", "Spacing", "Date"]
)["Calibrated_Height_m"].transform(lambda x: x.quantile(0.80))

df_dom = df_lmm[df_lmm["Calibrated_Height_m"] >= df_lmm["dom_threshold"]].copy()

print(f"\nAll trees:      {len(df_lmm):,} obs | {df_lmm['Tree_ID'].nunique():,} trees")
print(f"Dominant trees: {len(df_dom):,} obs | {df_dom['Tree_ID'].nunique():,} trees")

# ── 4. LMM formula builder ────────────────────────────────────────────────────
def build_formula(response):
    return f"{response} ~ Months * C(Species) + C(Spacing)"

def fit_lmm(data, response, label):
    print(f"\n  Fitting {label}...")
    formula = build_formula(response)
    try:
        model = smf.mixedlm(
            formula,
            data=data.dropna(subset=[response]),
            groups=data.dropna(subset=[response])["Plot_ID"],
            re_formula="~Months"
        )
        result = model.fit(reml=True, method="lbfgs", maxiter=200)
        print(f"    Converged: {result.converged} | "
              f"Log-lik: {result.llf:.1f} | "
              f"N obs: {result.nobs:,}")
        return result
    except Exception as e:
        print(f"    ERROR: {e}")
        return None

# ── 5. Fit all six models ─────────────────────────────────────────────────────
print("\n" + "="*60)
print("POPULATION 1: All alive trees")
print("="*60)
m_all_h     = fit_lmm(df_lmm, "Calibrated_Height_m", "Height — all trees")
m_all_crown = fit_lmm(df_lmm, "log_Crown",            "Crown Area (log) — all trees")
m_all_cah   = fit_lmm(df_lmm, "log_CAH",              "CA:H ratio (log) — all trees")

print("\n" + "="*60)
print("POPULATION 2: Dominant trees (top 20% height per group/date)")
print("="*60)
m_dom_h     = fit_lmm(df_dom, "Calibrated_Height_m", "Height — dominant trees")
m_dom_crown = fit_lmm(df_dom, "log_Crown",            "Crown Area (log) — dominant trees")
m_dom_cah   = fit_lmm(df_dom, "log_CAH",              "CA:H ratio (log) — dominant trees")

# ── 6. Extract clean summary tables ──────────────────────────────────────────
def extract_summary(result, model_label, response_label, log_scale=False):
    if result is None:
        return pd.DataFrame()

    params = result.params
    bse    = result.bse
    pvals  = result.pvalues
    ci     = result.conf_int()

    rows = []
    for term in params.index:
        if term in ["Group Var", "Group x Months Cov", "Months Var"]:
            continue
        row = {
            "Model":     model_label,
            "Response":  response_label,
            "Term":      term,
            "Coef":      round(params[term], 4),
            "SE":        round(bse[term], 4),
            "CI_lower":  round(ci.loc[term, 0], 4),
            "CI_upper":  round(ci.loc[term, 1], 4),
            "p_value":   round(pvals[term], 4),
            "Sig":       ("***" if pvals[term] < 0.001 else
                          "**"  if pvals[term] < 0.01  else
                          "*"   if pvals[term] < 0.05  else
                          "."   if pvals[term] < 0.10  else "ns"),
        }
        if log_scale:
            row["exp_coef"] = round(np.exp(params[term]), 4)
        rows.append(row)

    return pd.DataFrame(rows)

summary_parts = [
    extract_summary(m_all_h,     "All trees",  "Height (m)"),
    extract_summary(m_all_crown, "All trees",  "Crown Area (m²)",  log_scale=True),
    extract_summary(m_all_cah,   "All trees",  "CA:H ratio",       log_scale=True),
    extract_summary(m_dom_h,     "Dominant",   "Height (m)"),
    extract_summary(m_dom_crown, "Dominant",   "Crown Area (m²)",  log_scale=True),
    extract_summary(m_dom_cah,   "Dominant",   "CA:H ratio",       log_scale=True),
]

summary_df = pd.concat(summary_parts, ignore_index=True)
summary_df.to_csv("/content/drive/My Drive/EucVision/Figures/lmm_six_models_summary.csv", index=False)

# ── 7. FIGURE 1 — Coefficient plot comparing all six models ──────────────────
species_list = [s for s in df_lmm["Species"].cat.categories if s != REFERENCE_SPECIES]

# Inheriting colors dynamically from global template (Tab10)
pop_colors   = {"All trees": tab10_colors[0], "Dominant": tab10_colors[3]}
pop_markers  = {"All trees": "o",       "Dominant": "s"}
pop_labels   = {"All trees": "All trees", "Dominant": "Dominant (top 20%)"}

response_groups = [
    ("Height (m)",      [("All trees", m_all_h),   ("Dominant", m_dom_h)]),
    ("Crown Area (m²)", [("All trees", m_all_crown), ("Dominant", m_dom_crown)]),
    ("CA:H ratio",      [("All trees", m_all_cah),  ("Dominant", m_dom_cah)]),
]

fig, axes = plt.subplots(1, 3, figsize=(FW, 3.2), gridspec_kw={"wspace": 0.45})

for ax, (resp_label, model_pairs) in zip(axes, response_groups):
    y_positions = np.arange(len(species_list))
    offsets     = [-0.12, 0.12]

    for offset, (pop_label, result) in zip(offsets, model_pairs):
        if result is None:
            continue
        coefs, lowers, uppers = [], [], []
        for sp in species_list:
            term = f"Months:C(Species)[T.{sp}]"
            if term in result.params.index:
                c  = result.params[term]
                ci = result.conf_int().loc[term]
                coefs.append(c)
                lowers.append(ci[0])
                uppers.append(ci[1])
            else:
                coefs.append(np.nan)
                lowers.append(np.nan)
                uppers.append(np.nan)

        coefs  = np.array(coefs)
        lowers = np.array(lowers)
        uppers = np.array(uppers)

        y = y_positions + offset
        ax.errorbar(
            coefs, y,
            xerr=[coefs - lowers, uppers - coefs],
            fmt=pop_markers[pop_label],
            color=pop_colors[pop_label],
            markersize=4,
            capsize=2,
            linewidth=0.7,
            label=pop_labels[pop_label]
        )

    ax.axvline(0, color="grey", linewidth=0.6, linestyle="--")
    ax.set_yticks(y_positions)
    ax.set_yticklabels(species_list)

    xlabel_text = f"Growth rate difference\nvs {REFERENCE_SPECIES} per month"
    if "Crown" in resp_label or "CA:H" in resp_label:
        xlabel_text = "Growth rate difference\n(log scale) per month"

    ax.set_xlabel(xlabel_text)
    ax.set_title(resp_label)

# Shared legend
legend_elements = [
    Line2D([0], [0], marker="o", color="w", markerfacecolor=pop_colors["All trees"], markersize=5, label="All trees"),
    Line2D([0], [0], marker="s", color="w", markerfacecolor=pop_colors["Dominant"], markersize=5, label="Dominant (top 20%)"),
]
fig.legend(handles=legend_elements, loc="lower center", ncol=2, frameon=False, bbox_to_anchor=(0.5, -0.04))

fig.suptitle(
    f"Species growth rate differences relative to {REFERENCE_SPECIES}\n"
    f"(Months × Species interaction coefficients)", y=1.05
)

plt.savefig("/content/drive/My Drive/EucVision/Figures/lmm_coef_plot_species.png")
plt.show()

# ── 8. FIGURE 2 — Predicted growth curves: all trees vs dominant ──────────────
months_seq = np.linspace(0, df_lmm["Months"].max(), 200)

def predict_lmm_species(result, species_list, months_seq, reference_species, reference_spacing, log_scale=False):
    if result is None:
        return {}
    preds = {}
    params = result.params
    for sp in [reference_species] + species_list:
        y = np.zeros_like(months_seq)
        y += params.get("Intercept", 0)
        y += params.get(f"C(Species)[T.{sp}]", 0)
        y += params.get("Months", 0) * months_seq
        y += params.get(f"Months:C(Species)[T.{sp}]", 0) * months_seq
        if log_scale:
            y = np.exp(y)
        preds[sp] = (months_seq, y)
    return preds

fig, axes = plt.subplots(2, 3, figsize=(FW, 4.5), gridspec_kw={"hspace": 0.5, "wspace": 0.4})

row_labels   = ["All trees", "Dominant (top 20%)"]
col_labels   = ["Height (m)", "Crown area (m²)", "CA:H ratio (m² m⁻¹)"]
model_pairs  = [
    (m_all_h,     False), (m_all_crown, True), (m_all_cah,   True),
    (m_dom_h,     False), (m_dom_crown, True), (m_dom_cah,   True),
]
y_labels = ["Height (m)", "Crown area (m²)", "CA:H (m² m⁻¹)", "Height (m)", "Crown area (m²)", "CA:H (m² m⁻¹)"]

for idx, (ax, (result, log_scale), y_label) in enumerate(zip(axes.flat, model_pairs, y_labels)):
    preds = predict_lmm_species(
        result,
        species_list=[s for s in df_lmm["Species"].cat.categories if s != REFERENCE_SPECIES],
        months_seq=months_seq,
        reference_species=REFERENCE_SPECIES,
        reference_spacing=REFERENCE_SPACING,
        log_scale=log_scale
    )

    linestyle = "--" if idx < 3 else "-"

    for sp, (x, y) in preds.items():
        # Automatically pull from global species_colors dictionary
        color = species_colors.get(sp, "grey")
        ax.plot(x, y, color=color, linewidth=1.1, linestyle=linestyle, label=sp)

    row, col = idx // 3, idx % 3

    if row == 0: ax.set_title(col_labels[col])
    if col == 0: ax.set_ylabel(row_labels[row])
    if row == 1: ax.set_xlabel("Months from 1 Sep 2025")

    ax.set_ylabel(y_label)

# Shared legend
legend_handles = [
    Line2D([0], [0], color=species_colors.get(sp, "grey"), linewidth=1.1, label=sp)
    for sp in ([REFERENCE_SPECIES] + [s for s in df_lmm["Species"].cat.categories if s != REFERENCE_SPECIES])
    if sp in species_colors
]
legend_handles += [
    Line2D([0], [0], color="grey", linewidth=1.1, linestyle="--", label="All trees"),
    Line2D([0], [0], color="grey", linewidth=1.1, linestyle="-", label="Dominant (top 20%)"),
]
fig.legend(handles=legend_handles, loc="lower center", ncol=4, frameon=False, bbox_to_anchor=(0.5, -0.06))

fig.suptitle(
    "LMM-predicted growth trajectories by species\n"
    f"(Reference spacing: {REFERENCE_SPACING}×{REFERENCE_SPACING} m  |  Reference species: {REFERENCE_SPECIES})", y=1.05
)

plt.savefig("/content/drive/My Drive/EucVision/Figures/lmm_growth_curves_all_vs_dominant.png")
plt.show()

# ── 9. FIGURE 3 — Spacing effects: all six models ────────────────────────────
spacing_list = [s for s in df_lmm["Spacing"].cat.categories if s != REFERENCE_SPACING]

spacing_display = {1: "1×1 m", 2: "2×2 m", 3: "3×3 m", 5: "5×5 m"}

fig, axes = plt.subplots(2, 3, figsize=(FW, 4.0), gridspec_kw={"hspace": 0.55, "wspace": 0.5})

for idx, (ax, (result, log_scale)) in enumerate(zip(axes.flat, model_pairs)):
    if result is None:
        ax.set_visible(False)
        continue

    coefs, lowers, uppers, labels, colours = [], [], [], [], []

    # Replacing hardcoded hex with global tab10_colors for spacing bars
    spacing_colors_map = {1: tab10_colors[2], 2: tab10_colors[4], 3: tab10_colors[3], 5: tab10_colors[0]}

    for sc in spacing_list:
        term = f"C(Spacing)[T.{sc}]"
        if term in result.params.index:
            c  = result.params[term]
            ci = result.conf_int().loc[term]
            coefs.append(c)
            lowers.append(ci[0])
            uppers.append(ci[1])
            labels.append(spacing_display.get(sc, str(sc)))
            colours.append(spacing_colors_map.get(sc, "grey"))
        else:
            coefs.append(np.nan)
            lowers.append(np.nan)
            uppers.append(np.nan)
            labels.append(spacing_display.get(sc, str(sc)))
            colours.append("grey")

    coefs, lowers, uppers = np.array(coefs), np.array(lowers), np.array(uppers)
    y = np.arange(len(labels))

    for yi, (c, lo, hi, col) in enumerate(zip(coefs, lowers, uppers, colours)):
        if not np.isnan(c):
            ax.barh(yi, c, color=col, alpha=0.75, height=0.5)
            ax.errorbar(c, yi, xerr=[[c - lo], [hi - c]], fmt="none", color="0.3", linewidth=0.7, capsize=2)

    ax.axvline(0, color="grey", linewidth=0.6, linestyle="--")
    ax.set_yticks(y)
    ax.set_yticklabels(labels)

    row, col_idx = idx // 3, idx % 3

    if row == 0: ax.set_title(col_labels[col_idx])
    if col_idx == 0: ax.set_ylabel(row_labels[row])

    xlabel = (f"Coef. vs {REFERENCE_SPACING}×{REFERENCE_SPACING} m\n(log scale)"
              if log_scale else f"Coef. vs {REFERENCE_SPACING}×{REFERENCE_SPACING} m")
    ax.set_xlabel(xlabel)

fig.suptitle(
    f"Spacing effects (fixed intercept coefficients vs {REFERENCE_SPACING}×{REFERENCE_SPACING} m)\n"
    "Top row: all trees  |  Bottom row: dominant trees", y=1.05
)

plt.savefig("lmm_spacing_effects.pdf", dpi=300, bbox_inches="tight")
plt.show()

# ── 10. FIGURE 4 — Diagnostic Plots (Residuals vs. Fitted) ───────────────────
print("\nGenerating Diagnostic Plots...")

fig, axes = plt.subplots(2, 3, figsize=(FW, 5.0), gridspec_kw={"hspace": 0.4, "wspace": 0.4})

for idx, (ax, (result, log_scale), y_label) in enumerate(zip(axes.flat, model_pairs, y_labels)):
    if result is None:
        ax.set_visible(False)
        continue

    fitted = result.fittedvalues
    resids = result.resid

    # Scatter plot of residuals vs fitted values
    ax.scatter(fitted, resids, alpha=0.3, s=10, color="0.3", edgecolors="none")
    ax.axhline(0, color="red", linestyle="--", linewidth=1.2)

    row, col = idx // 3, idx % 3
    if row == 0: ax.set_title(col_labels[col])
    if col == 0: ax.set_ylabel(f"{row_labels[row]}\nResiduals")
    if row == 1: ax.set_xlabel("Fitted Values")

fig.suptitle(
    "Diagnostic: Residuals vs. Fitted Values\n(Checking for homoscedasticity and linearity)",
    y=1.05
)

plt.savefig("/content/drive/My Drive/EucVision/Figures/lmm_diagnostics_resid_vs_fitted.pdf")
plt.show()

# ── 11. FIGURE 5 — Diagnostic Plots (Q-Q Plots) ──────────────────────────────
fig, axes = plt.subplots(2, 3, figsize=(FW, 5.0), gridspec_kw={"hspace": 0.4, "wspace": 0.4})

for idx, (ax, (result, log_scale), y_label) in enumerate(zip(axes.flat, model_pairs, y_labels)):
    if result is None:
        ax.set_visible(False)
        continue

    resids = result.resid

    # Generate Q-Q plot
    stats.probplot(resids, dist="norm", plot=ax)

    # Clean up standard probplot styling to match your formatting
    ax.set_title("")

    # Adjust scatter points
    scatter_line = ax.get_lines()[0]
    scatter_line.set_marker('o')
    scatter_line.set_markerfacecolor('0.3')
    scatter_line.set_markeredgecolor('none')
    scatter_line.set_markersize(3)
    scatter_line.set_alpha(0.3)

    # Adjust theoretical line
    trend_line = ax.get_lines()[1]
    trend_line.set_color('red')
    trend_line.set_linewidth(1.2)
    trend_line.set_linestyle('--')

    row, col = idx // 3, idx % 3
    if row == 0: ax.set_title(col_labels[col])
    if col == 0: ax.set_ylabel(f"{row_labels[row]}\nOrdered Values")
    else: ax.set_ylabel("")

    if row == 1: ax.set_xlabel("Theoretical Quantiles")
    else: ax.set_xlabel("")

fig.suptitle(
    "Diagnostic: Normal Q-Q Plots of Residuals\n(Checking for normality of errors)",
    y=1.05
)

plt.savefig("/content/drive/My Drive/EucVision/Figures/lmm_diagnostics_qq.pdf")
plt.show()

# ── 12. FIGURE 6 — Diagnostic Plots (Residuals vs. Time) ─────────────────────
print("\nGenerating Residuals vs. Time (Months) Plots...")
import statsmodels.api as sm

fig, axes = plt.subplots(2, 3, figsize=(FW, 5.0), gridspec_kw={"hspace": 0.4, "wspace": 0.4})

for idx, (ax, (result, log_scale), y_label) in enumerate(zip(axes.flat, model_pairs, y_labels)):
    if result is None:
        ax.set_visible(False)
        continue

    # Extract residuals and the exact 'Months' values used in the fitted model
    resids = result.resid
    months = result.model.data.frame["Months"]

    # Scatter plot of the residual cloud
    ax.scatter(months, resids, alpha=0.3, s=10, color="0.3", edgecolors="none")

    # Baseline zero line
    ax.axhline(0, color="black", linestyle="--", linewidth=1.0)

    # Fit a LOWESS trend line to help visually detect temporal waves
    try:
        # frac=0.3 means the smoother looks at 30% of the data around each point
        lowess = sm.nonparametric.lowess(resids, months, frac=0.3)
        ax.plot(lowess[:, 0], lowess[:, 1], color="red", linewidth=1.5, linestyle="-")
    except Exception as e:
        print(f"Could not fit LOWESS for {y_label}: {e}")

    # Formatting
    row, col = idx // 3, idx % 3
    if row == 0: ax.set_title(col_labels[col])
    if col == 0: ax.set_ylabel(f"{row_labels[row]}\nResiduals")
    if row == 1: ax.set_xlabel("Time (Months)")

fig.suptitle(
    "Diagnostic: Residuals vs. Time (Months)\n(A wavy red LOWESS line indicates a GAMM might be needed)",
    y=1.05
)

plt.savefig("/content/drive/My Drive/EucVision/Figures/lmm_diagnostics_resid_vs_time.pdf")
plt.show()
```

```python
# @title Plot 37-40 Video

import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
import matplotlib.dates as mdates
from matplotlib.animation import FuncAnimation
from IPython.display import HTML
from statsmodels.nonparametric.smoothers_lowess import lowess

# --- 1. LOAD & MINIMAL PREPROCESSING ---
file_path = '/content/drive/My Drive/EucVision/01. Master Dataset.csv'
df_video_2 = pd.read_csv(file_path, low_memory=False)

# Clean, simple numerical date parsing
df_video_2['Date'] = pd.to_datetime(df_video_2['Date'], format='%d-%m-%Y', errors='coerce')

# Ensure critical measurement columns are treated as numeric
df_video_2['Crown_Area'] = pd.to_numeric(df_video_2['Crown_Area'], errors='coerce')
df_video_2['Plot'] = pd.to_numeric(df_video_2['Plot'], errors='coerce')

# --- 2. COLLAPSE SPLIT ROWS & TRACK UNIQUE TREES ---
# Generate unique identifier for each tree
df_video_2['Tree_ID'] = (df_video_2['Compartment'].astype(str) + '_' +
                 df_video_2['Line'].astype(str) + '_' +
                 df_video_2['Plot'].astype(str) + '_' +
                 df_video_2['Tree'].map('{:.2f}'.format))

# Collapse split rows before calculations happen
group_cols = ['Date', 'Compartment', 'Line', 'Plot', 'Culture', 'Spacing', 'Species', 'Tree', 'Tree_ID']
if 'Death_Date' in df_video_2.columns:
    group_cols.append('Death_Date')

df_video_2 = df_video_2.dropna(subset=['Date'])
df_video_2 = df_video_2.groupby(group_cols, dropna=False).first().reset_index()

# Sort chronologically by tree
df_video_2 = df_video_2.sort_values(by=['Tree_ID', 'Date'])

# --- 3. MORTALITY FILTERING ---
if 'Death_Date' in df_video_2.columns:
    df_video_2['Parsed_Death_Date'] = pd.to_datetime(df_video_2['Death_Date'], format='%d-%m-%Y', errors='coerce')
    df_video_2['Is_Dead'] = df_video_2['Parsed_Death_Date'].notna() & (df_video_2['Date'] >= df_video_2['Parsed_Death_Date'])

    # Force Crown_Area to NaN for dead trees so they don't skew the dominant means
    df_video_2.loc[df_video_2['Is_Dead'], 'Crown_Area'] = np.nan
    df_video_2 = df_video_2.drop(columns=['Parsed_Death_Date', 'Is_Dead'])

# --- 4. ISOLATE TARGET PLOTS ---
# Filter for Plots 37-40 safely
df_video_2 = df_video_2[df_video_2['Plot'].isin([37, 38, 39, 40])].copy()

# Treat Plot 39 as a single "Mixed" entity
df_video_2.loc[df_video_2['Plot'] == 39, 'Species'] = 'Mixed'

# --- 5. DOMINANT CROWN AREA CALCULATION ---
df_dom = df_video_2.dropna(subset=['Crown_Area', 'Species', 'Spacing']).copy()
results = []

for (date, spacing, species), group in df_dom.groupby(['Date', 'Spacing', 'Species']):
    spacing = int(spacing)

    # Isolate the Top 20% (Dominant) trees
    threshold = group['Crown_Area'].quantile(0.80)
    top_20 = group[group['Crown_Area'] >= threshold]
    dom_ca = top_20['Crown_Area'].mean()

    results.append({
        'Date': date,
        'Legend_Label': f"{species} ({spacing}m spacing)",
        'Dominant Crown Area': dom_ca
    })

res_df = pd.DataFrame(results)

# --- 6. PRE-CALCULATE LOESS SMOOTHING ---
res_df = res_df.sort_values(by=['Legend_Label', 'Date'])
res_df['Smoothed_Crown_Area'] = np.nan
smooth_frac = 0.3

for label, group in res_df.groupby('Legend_Label'):
    if len(group) > 3:
        x_num = mdates.date2num(group['Date'])
        y_val = group['Dominant Crown Area'].values

        # Calculate LOESS
        smoothed = lowess(y_val, x_num, frac=smooth_frac)
        # Apply the smoothed y-values back to the dataframe
        res_df.loc[group.index, 'Smoothed_Crown_Area'] = smoothed[:, 1]
    else:
        # Fallback to raw data if not enough points to smooth
        res_df.loc[group.index, 'Smoothed_Crown_Area'] = group['Dominant Crown Area']

# --- 7. SET UP CUSTOM COLOR PALETTE ---
unique_labels = res_df['Legend_Label'].unique()
custom_palette = {}

# Map the exact hex codes from your video
spacing_colors = {
    '1m': '#eedf13',
    '2m': '#f10070',
    '3m': '#15e94e',
    '5m': '#ff0004'
}

for label in unique_labels:
    # Assign color based on the spacing text found in the label
    if "1m spacing" in label:
        custom_palette[label] = spacing_colors['1m']
    elif "2m spacing" in label:
        custom_palette[label] = spacing_colors['2m']
    elif "3m spacing" in label:
        custom_palette[label] = spacing_colors['3m']
    elif "5m spacing" in label:
        custom_palette[label] = spacing_colors['5m']
    else:
        custom_palette[label] = '#000000' # Fallback black just in case

# --- 8. ANIMATION SETUP ---
# Implemented FW and single panel height (4.0), plus standardizing to 300 DPI
fig, ax = plt.subplots(figsize=(FW, 4.0), dpi=300)
unique_dates = sorted(res_df['Date'].unique())

# Fix the Y-axis limit using the max of both smoothed and raw data
max_y = max(res_df['Dominant Crown Area'].max(), res_df['Smoothed_Crown_Area'].max()) * 1.05

date_buffer = pd.Timedelta(days=15)
min_x = min(unique_dates) - date_buffer
max_x = max(unique_dates) + date_buffer

def update(frame_date):
    ax.clear()
    current_data = res_df[res_df['Date'] <= frame_date]

    # 1. Plot Faint Raw Data (Markers and thin lines)
    sns.lineplot(
        data=current_data, x='Date', y='Dominant Crown Area',
        hue='Legend_Label', palette=custom_palette,
        marker='o', linestyle='-', linewidth=0.8, alpha=0.3, # Scaled down for FW canvas
        legend=False, ax=ax # Legend handled by the smoothed line
    )

    # 2. Plot Bold Smoothed Trend
    sns.lineplot(
        data=current_data, x='Date', y='Smoothed_Crown_Area',
        hue='Legend_Label', palette=custom_palette,
        linewidth=1.5, # Scaled down for FW canvas
        legend=True, ax=ax
    )

    # Formatting: Removed hardcoded fonts and paddings to use rcParams!
    ax.set_title("Mean Crown Area (Plot 37 - 40)")
    ax.set_xlim(min_x, max_x)
    ax.set_ylim(0, max_y)
    ax.set_xlabel('Date')
    ax.set_ylabel('Crown Area (m²)')

    ax.xaxis.set_major_formatter(mdates.DateFormatter('%b %Y'))
    ax.xaxis.set_major_locator(mdates.MonthLocator(interval=2))
    ax.tick_params(axis='x', rotation=45)

    # Clean up duplicate legends from plotting multiple times
    if ax.get_legend() is not None:
        handles, labels = ax.get_legend_handles_labels()
        by_label = dict(zip(labels, handles))
        # Removed font kwargs; keeping title and inheriting the rest
        ax.legend(by_label.values(), by_label.keys(), title='Species & Spacing')

# Adjusted margins for the 4.0-inch height to prevent date cutoff
plt.subplots_adjust(bottom=0.25, left=0.10, right=0.95, top=0.90)

# --- 9. RENDER & SAVE EXPORT ---
# Each frame stays on screen for 270 milliseconds to match the R Script
ani = FuncAnimation(fig, update, frames=unique_dates, interval=270, repeat=False)
plt.close()

# Save the animation as an MP4 video
output_path = '/content/drive/My Drive/EucVision/Plot37_40_Dashboard_Animation_FW.mp4'

# Implemented 300 DPI for high-res render at 1890x1200 pixels
ani.save(output_path, writer='ffmpeg', fps=3.7, dpi=300)
print(f"Animation successfully saved to: {output_path}")
```
